### 🎓 1.1 · El problema que resuelven las funciones

Imagina que eres el analista de una universidad y necesitas calcular la **nota final** de tres estudiantes. Con lo que sabes hasta ahora, escribirías algo así:

```python
# Estudiante 1
nota_final_ana = (3.8 * 0.4) + (4.2 * 0.6)
print("Ana:", nota_final_ana)

# Estudiante 2
nota_final_luis = (2.5 * 0.4) + (3.1 * 0.6)
print("Luis:", nota_final_luis)

# Estudiante 3
nota_final_maria = (4.5 * 0.4) + (4.8 * 0.6)
print("María:", nota_final_maria)
```

Funciona... pero tiene **tres problemas graves**:

| Problema | Por qué importa |
|---|---|
| 🔁 **Repetición** | La misma fórmula escrita 3 veces. Con 500 estudiantes serían 500 líneas |
| 🐛 **Fragilidad** | Si la universidad cambia los pesos a 30/70, hay que corregir **cada** línea. Si olvidas una, tendrás notas mal calculadas y nadie se dará cuenta |
| 😵 **Ilegibilidad** | Al leerlo dentro de un mes, tendrás que descifrar qué significan `0.4` y `0.6` |

Este problema es tan común que tiene un nombre en programación: se resuelve con el principio **DRY** — *Don't Repeat Yourself*, "no te repitas".

> 🍳 **La analogía:** una función es como una **receta**. No le explicas a alguien cómo hacer un huevo frito cada vez que tiene hambre: escribes la receta **una vez**, le pones un nombre ("huevo frito") y desde entonces solo dices *"hazme un huevo frito"*. Si algún día mejoras la receta, la corriges en un solo lugar.

### 🎓 1.2 · La definición

> Una **función** es un bloque de código **reutilizable** que realiza una tarea específica. Ayuda a **estructurar** el programa, mejorar la **legibilidad** y **evitar la duplicación** de código.

Una función tiene tres momentos, y confundirlos es el error #1 de los principiantes:

| Momento | Qué pasa | Ejemplo |
|:---:|---|---|
| 1. **Definir** | Escribes la receta. **No se ejecuta nada todavía** | `def saludar(nombre):` |
| 2. **Llamar** | Pides que se ejecute, dándole los datos que necesita | `saludar("Ana")` |
| 3. **Devolver** | La función te entrega su resultado | `return nota_final` |

### 🔬 1.3 · Anatomía de una función

Antes de ejecutar nada, vamos a desarmar la sintaxis pieza por pieza:

```python
def  calcular_nota_final (corte1, corte2):
│    │                    │
│    │                    └── PARÁMETROS: los datos que la función necesita recibir
│    └── NOMBRE: cómo la vas a llamar después
└── def: palabra reservada que dice "aquí empieza una función"

    """Calcula la nota final ponderada."""     ← DOCSTRING: qué hace (documentación)
    
    resultado = (corte1 * 0.4) + (corte2 * 0.6)   ← CUERPO: las operaciones
    
    return resultado                              ← RETURN: qué entrega de vuelta
```

| Pieza | Regla que debes recordar |
|---|---|
| **`def`** | Siempre en minúscula. Es una palabra reservada de Python |
| **Nombre** | En `snake_case`, sin espacios ni tildes, y **empezando por un verbo**: `calcular_`, `contar_`, `limpiar_`, `es_` |
| **`( )`** | Van siempre, **aunque la función no reciba nada**: `def mostrar_menu():` |
| **`:`** | Los dos puntos al final de la línea abren el bloque. **Olvidarlos es un `SyntaxError`** |
| **Sangría** | Los **4 espacios** al inicio de cada línea del cuerpo **no son decoración**: son lo que le dice a Python qué está dentro de la función |
| **Docstring** | Texto entre `"""comillas triples"""` justo después del `def`. No es obligatorio, pero es lo que separa a un aficionado de un profesional |
| **`return`** | Entrega el resultado. **Si no lo escribes, la función devuelve `None`** |

````python
# ▶️ EJECUTA ESTA CELDA
# Fíjate: primero se DEFINE la función (no pasa nada visible),
# y solo cuando se LLAMA es que aparece el resultado.

def saludar(nombre):
    """Imprime un saludo personalizado para la persona indicada."""
    print(f"¡Hola, {nombre}! Bienvenido a DataXperience.")


print("1. La función ya está definida (pero todavía no ha hecho nada).")
print()

# Ahora la LLAMAMOS tres veces, con datos distintos:
saludar("Ana")
saludar("Luis")
saludar("María")
````

### 🧐 Lo que acabas de ver

En tu pantalla debió aparecer:

```
1. La función ya está definida (pero todavía no ha hecho nada).

¡Hola, Ana! Bienvenido a DataXperience.
¡Hola, Luis! Bienvenido a DataXperience.
¡Hola, María! Bienvenido a DataXperience.
```

Tres cosas para notar:

1. **La definición no imprimió nada.** Escribir `def saludar(...)` solo *guarda* la receta en la memoria de Python. Es como tener el libro de recetas cerrado en la cocina.
2. **Una sola definición, tres usos.** Escribiste la lógica del saludo **una vez** y la usaste **tres veces** con datos distintos. Ese es exactamente el punto de las funciones.
3. **`nombre` cambió de valor en cada llamada.** Cuando escribes `saludar("Ana")`, la palabra `"Ana"` entra a la función y ocupa el lugar de `nombre`. Ese hueco que se llena se llama **parámetro**.

> 💡 **La `f` antes de las comillas** en `f"¡Hola, {nombre}!"` se llama *f-string*. Le dice a Python: *"dentro de este texto, reemplaza lo que esté entre llaves `{}` por su valor real"*. Sin la `f`, imprimiría literalmente `¡Hola, {nombre}!`.

### ⚠️ Errores comunes en esta parte

Estos son los tres errores que casi con seguridad vas a cometer. Aprende a **leer el mensaje**, porque te dice exactamente qué pasó:

| Si ves esto... | Significa que... | Se arregla así |
|---|---|---|
| `SyntaxError: expected ':'` | Olvidaste los **dos puntos** al final del `def` | `def saludar(nombre):` |
| `IndentationError: expected an indented block` | El cuerpo de la función **no tiene sangría** | Presiona `Tab` al inicio de las líneas del cuerpo |
| `NameError: name 'saludar' is not defined` | Llamaste la función **antes de definirla**, o no ejecutaste la celda donde está el `def` | Ejecuta primero la celda de la definición |
| `TypeError: saludar() missing 1 required positional argument: 'nombre'` | Llamaste la función **sin darle** el dato que pide | `saludar("Ana")` en vez de `saludar()` |

> 🧘 **Los errores no son fracasos, son mensajes.** Python te está diciendo con precisión qué le falta. Léelos de abajo hacia arriba: la última línea es siempre la más informativa.

### 🎓 1.4 · `return` vs `print`: la confusión más importante del curso

Esta distinción confunde a **todo el mundo** al principio, y es la causa de la mitad de los errores de un principiante. Léela con calma.

| | `print()` | `return` |
|---|---|---|
| **Qué hace** | **Muestra** algo en la pantalla | **Devuelve** un valor al programa |
| **¿Se puede guardar?** | ❌ No: obtienes `None` | ✅ Sí: se guarda en una variable |
| **¿Se puede seguir calculando?** | ❌ No | ✅ Sí |
| **Le habla a...** | El **humano** que mira la pantalla | El **resto de tu código** |

> 📮 **La analogía:** imagina que le pides a alguien que calcule algo.
> - Con **`print`** es como si lo **gritara en voz alta** y se fuera. Lo escuchaste, pero no tienes nada en la mano.
> - Con **`return`** es como si te lo **entregara escrito en un papel**. Ahora puedes guardarlo, sumarlo, compararlo o pasárselo a otra persona.

**La regla práctica:** si el resultado de tu función lo vas a **necesitar después**, usa `return`. `print` es solo para mostrar.

````python
# ▶️ EJECUTA ESTA CELDA
# La MISMA tarea escrita de dos formas. Observa la diferencia al intentar guardar el resultado.

def promedio_con_print(notas):
    """Muestra el promedio en pantalla, pero NO lo devuelve."""
    print(sum(notas) / len(notas))


def promedio_con_return(notas):
    """Calcula el promedio y lo DEVUELVE para poder usarlo después."""
    return sum(notas) / len(notas)


mis_notas = [4.0, 3.5, 4.5, 3.0]

print("--- Llamando la versión con print ---")
resultado_a = promedio_con_print(mis_notas)   # imprime el número...

print("\n--- Llamando la versión con return ---")
resultado_b = promedio_con_return(mis_notas)  # ...no imprime nada visible

print("\n--- ¿Qué quedó guardado en cada variable? ---")
print("resultado_a (de print) :", resultado_a)
print("resultado_b (de return):", resultado_b)
````

### 🧐 Lo que acabas de ver

```
--- Llamando la versión con print ---
3.75

--- Llamando la versión con return ---

--- ¿Qué quedó guardado en cada variable? ---
resultado_a (de print) : None
resultado_b (de return): 3.75
```

**Esto es lo crucial:**

- La versión con `print` **mostró** `3.75` en pantalla, pero cuando intentaste guardarlo, `resultado_a` quedó valiendo **`None`** (= nada). El número se "gritó" y se perdió.
- La versión con `return` **no mostró nada** al llamarla (por eso la segunda sección se ve vacía), pero `resultado_b` sí **contiene** el `3.75`.

Y eso cambia todo lo que puedes hacer después: sobre `resultado_b` puedes seguir operando. Sobre `resultado_a`, no.

````python
# ▶️ EJECUTA ESTA CELDA
# La prueba definitiva: con el valor devuelto por return SÍ puedo seguir trabajando.

promedio = promedio_con_return(mis_notas)

print("El promedio es:", promedio)
print("¿Aprobó (>= 3.0)?:", promedio >= 3.0)
print("Sobre 100:", promedio * 20)
print("Redondeado a 1 decimal:", round(promedio, 1))

# Nada de esto sería posible con la versión de print:
# resultado_a >= 3.0   ->  TypeError, porque None no se puede comparar con un número
````

### 🎓 1.5 · Parámetros: valores por defecto y retorno múltiple

Hay dos capacidades más de las funciones que usarás todo el tiempo en análisis de datos:

#### A) Parámetros por defecto

A veces un dato **casi siempre vale lo mismo**. En Colombia el IVA es del 19 %... salvo para algunos alimentos, que es del 5 %. En lugar de escribir dos funciones, le das un **valor por defecto** al parámetro:

```python
def calcular_precio(monto, iva=0.19):
```

Ahora `iva` es **opcional**: si no lo pasas, Python usa `0.19`; si lo pasas, usa el tuyo.

> ⚠️ **Regla obligatoria:** los parámetros con valor por defecto van **siempre al final**. `def f(a=1, b)` es un error de sintaxis.

#### B) Devolver varios valores

Una función puede devolver **más de un resultado** separándolos con comas. Python los empaqueta en una tupla y tú los puedes "desempacar" en varias variables:

```python
return nota_final, aprobo        # devuelve dos cosas

nota, paso = calcular_nota(...)  # las recibe en dos variables
```

Esto es utilísimo: una función de análisis suele necesitar entregar **el resultado y su interpretación** a la vez.

````python
# ▶️ EJECUTA ESTA CELDA

def calcular_nota_final(corte1, corte2, peso_corte1=0.4):
    """Calcula la nota final ponderada de un estudiante.

    Parámetros:
        corte1 (float): nota del primer corte, de 0 a 5
        corte2 (float): nota del segundo corte, de 0 a 5
        peso_corte1 (float): peso del primer corte. Por defecto 0.4 (40%)

    Devuelve:
        (float, bool): la nota final y si aprobó (>= 3.0)
    """
    peso_corte2 = 1 - peso_corte1                                  # si el 1º pesa 40%, el 2º pesa 60%
    nota_final = (corte1 * peso_corte1) + (corte2 * peso_corte2)   # 🥇 paréntesis por claridad
    aprobo = nota_final >= 3.0                                     # esto da True o False
    return nota_final, aprobo                                      # devuelve DOS valores


# Caso 1: usando el peso por defecto (40 % / 60 %)
nota, paso = calcular_nota_final(3.8, 4.2)
print(f"Pesos 40/60 -> nota final: {nota:.2f} | ¿Aprobó?: {paso}")

# Caso 2: cambiando el peso a 50 % / 50 %
nota, paso = calcular_nota_final(3.8, 4.2, peso_corte1=0.5)
print(f"Pesos 50/50 -> nota final: {nota:.2f} | ¿Aprobó?: {paso}")

# Caso 3: un estudiante que no alcanzó
nota, paso = calcular_nota_final(2.0, 2.5)
print(f"Pesos 40/60 -> nota final: {nota:.2f} | ¿Aprobó?: {paso}")
````

### 🧐 Lo que acabas de ver

```
Pesos 40/60 -> nota final: 4.04 | ¿Aprobó?: True
Pesos 50/50 -> nota final: 4.00 | ¿Aprobó?: True
Pesos 40/60 -> nota final: 2.30 | ¿Aprobó?: False
```

Cuatro detalles que vale la pena entender:

1. **En el caso 1 no escribiste `peso_corte1`** y aun así funcionó: Python usó el valor por defecto `0.4`. Eso es lo que hace opcional a un parámetro.
2. **En el caso 2 escribiste `peso_corte1=0.5`.** Nombrar el parámetro al llamarlo se llama *argumento con nombre* y hace tu código **mucho más legible**: cualquiera entiende qué es ese `0.5`.
3. **La función devolvió dos valores** y tú los recibiste en dos variables (`nota, paso`). El orden importa: llegan en el mismo orden en que están en el `return`.
4. **`{nota:.2f}`** dentro del f-string significa "muestra este número con 2 decimales". Es formato de presentación, no cambia el valor guardado.

### 🧐 Lo que acabas de ver

Python te mostró **exactamente el docstring que escribiste**. Esto es enorme por dos razones:

- Dentro de seis meses, cuando no recuerdes cómo usar tu propia función, `help(mi_funcion)` te lo dirá.
- Es el **mismo mecanismo** que usan las librerías profesionales. Prueba a ejecutar `help(sum)` o `help(len)` en cualquier celda y verás la documentación oficial de Python, escrita igualito.

> 🏅 **Buena práctica:** un docstring debería responder tres cosas: **qué hace** la función, **qué recibe** y **qué devuelve**.

```python
def calcular_precio_final(precio, iva=0.19):

    Parámetros:
```

### 🎓 3.1 · Qué es el control de flujo

Hasta ahora tus programas se ejecutaban **de arriba abajo**, línea por línea, siempre igual. Eso se llama ejecución **secuencial**.

Pero los programas útiles necesitan **reaccionar**: aprobar o reprobar según la nota, alertar o no según la temperatura, aceptar o rechazar según el dato. Eso se llama **control de flujo**: decidir **en qué orden** —y si acaso— se ejecutan las instrucciones.

Estas son todas las herramientas de control de flujo de Python. Las verás **todas** en este cuaderno:

| Estructura | Para qué sirve | Ejemplo | Parte |
|:---|:---|:---|:---:|
| `if` | Ejecutar **si** se cumple una condición | `if x > 0:` | 3 |
| `elif` | Probar **otra** condición si la anterior falló | `elif x == 0:` | 3 |
| `else` | Qué hacer si **nada** se cumplió | `else:` | 3 |
| `for` | **Recorrer** una secuencia elemento por elemento | `for i in lista:` | 4 |
| `while` | Repetir **mientras** una condición sea verdadera | `while x < 10:` | 4 |
| `break` | **Salir** del bucle inmediatamente | `if x == 5: break` | 5 |
| `continue` | **Saltar** a la siguiente vuelta del bucle | `if x == 3: continue` | 5 |

### 🔬 3.2 · Anatomía de un condicional

```python
if  nota >= 3.0 :
│   └────┬────┘ └── los dos puntos abren el bloque
│        └── LA CONDICIÓN: cualquier expresión que dé True o False
└── la palabra clave

    print("Aprobó")     ← este bloque (con sangría) se ejecuta SOLO si la condición fue True

else :
    print("Reprobó")    ← este bloque se ejecuta SOLO si fue False
```

```
              ¿nota >= 3.0?
             /            \
          True            False
            ↓               ↓
     print("Aprobó")   print("Reprobó")
```

> 💡 **La condición no es magia:** es exactamente una de las comparaciones que ya conoces (`>`, `<`, `==`, `!=`, `>=`, `<=`) o una combinación de ellas con `and`, `or`, `not`. Lo único que hace `if` es preguntar: *"¿esto es `True`?"*

````python
# ▶️ EJECUTA ESTA CELDA
# Un condicional con tres caminos: if -> elif -> else

def clasificar_nota(nota):
    """Devuelve la categoría cualitativa de una nota entre 0 y 5.

    Parámetros:
        nota (float): la nota numérica

    Devuelve:
        str: la categoría correspondiente
    """
    if nota >= 4.5:                    # se evalúa PRIMERO
        return "Excelente 🏆"
    elif nota >= 4.0:                  # solo si la anterior fue False
        return "Sobresaliente 🌟"
    elif nota >= 3.0:                  # solo si las dos anteriores fueron False
        return "Aprobado ✅"
    else:                              # si NINGUNA se cumplió
        return "Reprobado ❌"


# Probamos la función con cuatro notas distintas:
for n in [4.8, 4.1, 3.2, 2.4]:
    print(n, "->", clasificar_nota(n))
````

### 🧐 Lo que acabas de ver

```
4.8 -> Excelente 🏆
4.1 -> Sobresaliente 🌟
3.2 -> Aprobado ✅
2.4 -> Reprobado ❌
```

**Cómo funciona por dentro, para la nota 4.1:**

1. Python pregunta: *¿4.1 >= 4.5?* → **False**. Se salta ese bloque.
2. Pregunta: *¿4.1 >= 4.0?* → **True**. ✅ Ejecuta `return "Sobresaliente"` **y se detiene ahí mismo**.
3. Los `elif` y el `else` que quedaban **ni siquiera se evalúan**.

> 🔑 **La regla de oro de los condicionales:** Python evalúa **de arriba abajo** y se detiene en la **primera** condición verdadera. Todo lo que venga después se ignora.

Esa regla tiene una consecuencia enorme, y es la fuente de un error clásico. Míralo en la siguiente celda.

````python
# ▶️ EJECUTA ESTA CELDA
# La MISMA lógica, pero con las condiciones en el ORDEN EQUIVOCADO.

def clasificar_mal(nota):
    """Versión con las condiciones mal ordenadas. ¡Mira lo que pasa!"""
    if nota >= 3.0:              # ⚠️ esta condición es demasiado 'ancha' y va primero
        return "Aprobado"
    elif nota >= 4.5:            # 💀 esta línea NUNCA se alcanza
        return "Excelente"
    else:
        return "Reprobado"


print("Con una nota de 4.8...")
print("  La versión mal ordenada dice:", clasificar_mal(4.8))
print("  La versión correcta dice    :", clasificar_nota(4.8))
````

### 🧐 Lo que acabas de ver

```
Con una nota de 4.8...
  La versión mal ordenada dice: Aprobado
  La versión correcta dice    : Excelente 🏆
```

**Un 4.8 clasificado como simple "Aprobado".** ¿Por qué? Porque `4.8 >= 3.0` **también es verdad**, y como esa condición estaba primero, Python entró ahí y nunca llegó a preguntar por el 4.5.

La línea `elif nota >= 4.5` es **código muerto**: está escrita, se ve bien, y jamás se ejecutará.

<b style="color:#C0392B">⚠️ Y lo peor: Python no da ningún error.</b><br>
El programa corre perfecto y entrega un resultado equivocado. Es el mismo tipo de error silencioso que viste con la jerarquía de operaciones en la clase pasada.
</div>

> 🥇 **Regla práctica:** cuando encadenes condiciones sobre un mismo número, ordénalas **de la más restrictiva a la más amplia** (o de mayor a menor). Y siempre pruébalas con un caso de cada categoría.

### 🎓 3.3 · Condiciones combinadas: aquí regresan `and`, `or` y `not`

Las reglas de la vida real casi nunca dependen de una sola cosa. *"Aprueba la beca si tiene buen promedio **y** va en tercer semestre o más **y** no ha perdido materias"*.

Para eso están los **operadores lógicos** que viste en la clase anterior:

| Operador | Devuelve `True` cuando… | Ejemplo |
|---|---|---|
| `and` | **ambas** condiciones son verdaderas | `promedio >= 4.0 and semestre >= 3` |
| `or` | **al menos una** es verdadera | `semestre == 1 or estrato <= 2` |
| `not` | invierte el resultado | `not aprobo` |

> 🥇 **Regla de Oro (otra vez):** cuando mezcles `and` con `or`, **usa paréntesis**. Python evalúa `not` primero, luego `and`, y `or` de último — pero nadie debería tener que recordar eso para leer tu código.

💡 **Truco de legibilidad profesional:** en vez de escribir una condición gigante dentro del `if`, guárdala en **variables con nombres descriptivos**. Compara:

```python
# Difícil de leer 😵
if promedio >= 4.0 and materias_perdidas == 0 and semestre >= 3:

# Fácil de leer ✅
cumple_academico = promedio >= 4.0 and materias_perdidas == 0
cumple_avance = semestre >= 3
if cumple_academico and cumple_avance:
```

La segunda versión **se lee casi como español**.

````python
# ▶️ EJECUTA ESTA CELDA

def evaluar_beca(promedio, semestre, materias_perdidas):
    """Determina si un estudiante cumple los requisitos de la beca DataXperience.

    Requisitos: promedio >= 4.0, sin materias perdidas y semestre 3 o superior.

    Parámetros:
        promedio (float): promedio acumulado
        semestre (int): semestre que cursa
        materias_perdidas (int): materias perdidas este semestre

    Devuelve:
        str: el veredicto con su explicación
    """
    # Guardamos cada requisito en una variable con nombre claro:
    cumple_academico = promedio >= 4.0 and materias_perdidas == 0
    cumple_avance = semestre >= 3

    # Y ahora el condicional se lee solo:
    if cumple_academico and cumple_avance:
        return "🎓 Beca aprobada"
    elif cumple_academico and not cumple_avance:
        return "⏳ En espera: buen desempeño, pero faltan semestres"
    else:
        return "❌ No cumple los requisitos académicos"


print(evaluar_beca(promedio=4.3, semestre=5, materias_perdidas=0))
print(evaluar_beca(promedio=4.3, semestre=2, materias_perdidas=0))
print(evaluar_beca(promedio=3.5, semestre=6, materias_perdidas=1))
````

### 🧐 Lo que acabas de ver

```
🎓 Beca aprobada
⏳ En espera: buen desempeño, pero faltan semestres
❌ No cumple los requisitos académicos
```

**Analicemos el segundo caso** (promedio 4.3, semestre 2, 0 materias perdidas):

1. `cumple_academico` = `4.3 >= 4.0` **and** `0 == 0` → `True and True` → **`True`**
2. `cumple_avance` = `2 >= 3` → **`False`**
3. Primer `if`: `True and False` → `False`. No entra.
4. Primer `elif`: `True and not False` → `True and True` → **`True`**. ✅ Entra aquí.

Fíjate también en que las llamadas usan **argumentos con nombre** (`promedio=4.3`). No es obligatorio, pero cuando una función recibe tres números seguidos, nombrarlos evita el error clásico de pasarlos en el orden equivocado.

```python
print(notas[1])
print(notas[2])
print(notas[3])
print(notas[4])
```

Funciona. Pero ¿y si son 500 notas? ¿Y si no sabes cuántas hay hasta que llega el archivo?

Un **ciclo** (o bucle) le dice a Python: *"repite estas instrucciones para cada elemento, sean los que sean"*. Es, sin exagerar, **la herramienta más importante de la programación aplicada a datos**: un dataset no es más que muchas filas sobre las que hay que hacer lo mismo.

### 🎓 4.2 · `for` vs `while`: cuál usar

| | `for` | `while` |
|---|---|---|
| **Se usa cuando** | **Sabes sobre qué recorrer**: una lista, un rango de números, un diccionario | **No sabes cuántas veces**: dependes de que se cumpla una condición |
| **Se lee como** | "**para cada** elemento de la lista…" | "**mientras** se cumpla que…" |
| **Cuándo termina** | Cuando se acaban los elementos (garantizado) | Cuando la condición se vuelve falsa (**no garantizado**) |
| **Riesgo** | Prácticamente ninguno | **Bucle infinito** si la condición nunca cambia |
| **Ejemplo en datos** | Recorrer los registros de un archivo | Reintentar una descarga hasta que funcione |

> 🎯 **En ciencia de datos usarás `for` el 90 % del tiempo.** Recorrer registros es la operación más común del oficio.

### 🔬 4.3 · Anatomía de un `for`

```python
for  nota  in  notas :
│    │     │   │     └── los dos puntos abren el bloque
│    │     │   └── LA COLECCIÓN que vamos a recorrer
│    │     └── palabra clave obligatoria
│    └── LA VARIABLE que va tomando el valor de cada elemento, uno por uno
└── palabra clave

    print(nota)      ← el cuerpo: se ejecuta UNA VEZ POR CADA elemento
```

**Lo que hace Python por dentro**, si `notas = [4.0, 3.5, 2.8]`:

| Vuelta | `nota` vale | Se ejecuta |
|:---:|:---:|---|
| 1ª | `4.0` | `print(4.0)` |
| 2ª | `3.5` | `print(3.5)` |
| 3ª | `2.8` | `print(2.8)` |
| — | *se acabaron* | el ciclo termina |

> 💡 **El nombre de la variable lo eliges tú.** `for nota in notas` se lee mejor que `for x in notas`. Costumbre útil: la colección en plural, la variable en singular.

### 🔬 4.4 · `range()`: cuando quieres repetir un número de veces

Si no tienes una lista sino que solo quieres repetir *N* veces, usas `range()`:

| Escribes | Genera | Se lee |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | "cinco números empezando en 0" |
| `range(1, 6)` | 1, 2, 3, 4, 5 | "desde 1 hasta **antes de** 6" |
| `range(0, 21, 5)` | 0, 5, 10, 15, 20 | "de 0 a 20, de 5 en 5" |

> ⚠️ **`range` nunca incluye el último número.** `range(1, 6)` llega hasta 5. Es la misma lógica del *slicing* de listas que viste en la clase pasada. Por eso, para contar del 1 al 10 se escribe `range(1, 11)`.

````python
# ▶️ EJECUTA ESTA CELDA

def contar_hasta(limite):
    """Imprime los números desde 1 hasta el límite indicado (incluido)."""
    print(f"Contando hasta {limite}:")
    for i in range(1, limite + 1):     # el +1 es para que SÍ incluya el límite
        print(" ", i)


contar_hasta(5)

print("\n--- Las tres formas de range() ---")
print("range(5)       ->", list(range(5)))
print("range(2, 8)    ->", list(range(2, 8)))
print("range(0, 21, 5)->", list(range(0, 21, 5)))
````


**Fíjate en el `limite + 1`** dentro de `range(1, limite + 1)`. Si hubieras escrito `range(1, limite)`, el conteo habría llegado solo hasta 4. Ese "+1" es la corrección más frecuente que harás al usar `range`.

**Y el `list(...)`** que envuelve a `range` en las últimas líneas: `range` no es una lista, es un generador perezoso que produce los números a medida que se necesitan (por eficiencia). `list()` lo obliga a mostrarlos todos de una vez, solo para que puedas verlos.

````python
# ▶️ EJECUTA ESTA CELDA
# Las tres formas de recorrer que usarás en datos

notas = [4.0, 3.5, 2.8, 4.7, 3.1]

# --- FORMA 1: recorrer los valores de una lista ---
print("1) Sumando las notas una por una:")
suma = 0
for nota in notas:
    suma = suma + nota          # el acumulador crece en cada vuelta
    print(f"   sumé {nota} -> llevo {suma:.1f}")
print(f"   Promedio final: {suma / len(notas):.2f}")

# --- FORMA 2: recorrer con la posición, usando enumerate() ---
print("\n2) Con el número de estudiante:")
for posicion, nota in enumerate(notas, start=1):
    print(f"   Estudiante {posicion}: {nota}")

# --- FORMA 3: recorrer un diccionario (clave y valor a la vez) ---
print("\n3) Recorriendo un diccionario:")
inventario = {"café": 40, "arepa": 25, "jugo": 15}
for producto, cantidad in inventario.items():
    print(f"   {producto:<8} -> {cantidad} unidades")
````

### 🎓 4.5 · El combo más poderoso: `for` + `if`

Aquí es donde todo lo anterior se une y empieza a parecerse al trabajo real de un analista.

Un ciclo `for` recorre **todos** los registros; un `if` adentro decide **cuáles** te interesan. Con esas dos líneas puedes:

- **filtrar** (quedarte con los que cumplen algo),
- **contar** (cuántos cumplen),
- **acumular** (sumar solo los que cumplen),
- **clasificar** (separarlos en grupos).

Es, literalmente, lo que hace un `WHERE` en una base de datos o un filtro en Excel — solo que tú lo estás construyendo.

````python
# ▶️ EJECUTA ESTA CELDA
# Un mini-dataset: lista de diccionarios (igual que en la clase pasada)

registros = [
    {"nombre": "Ana",   "edad": 20, "promedio": 4.3},
    {"nombre": "Luis",  "edad": 22, "promedio": 2.9},
    {"nombre": "María", "edad": 21, "promedio": 4.8},
    {"nombre": "Juan",  "edad": 23, "promedio": 3.4},
]

# --- FILTRAR: mostrar solo los que cumplen una condición ---
print("🌟 Estudiantes con promedio superior a 4.0:")
for estudiante in registros:
    if estudiante["promedio"] > 4.0:
        print("   -", estudiante["nombre"], estudiante["promedio"])

# --- ACUMULAR EN UNA LISTA NUEVA: el patrón más usado en limpieza de datos ---
destacados = []                                  # 1. lista vacía ANTES del ciclo
for estudiante in registros:                     # 2. recorrer todo
    if estudiante["promedio"] > 4.0:             # 3. decidir
        destacados.append(estudiante["nombre"])  # 4. guardar si cumple

print("\n📋 Lista de destacados:", destacados)

# --- CONTAR Y PROMEDIAR ---
total_promedios = 0
cantidad_aprobados = 0
for estudiante in registros:
    total_promedios = total_promedios + estudiante["promedio"]
    if estudiante["promedio"] >= 3.0:
        cantidad_aprobados = cantidad_aprobados + 1

print(f"\n📊 Promedio del grupo : {total_promedios / len(registros):.2f}")
print(f"📊 Aprobados          : {cantidad_aprobados} de {len(registros)}")
print(f"📊 Tasa de aprobación : {(cantidad_aprobados / len(registros)) * 100:.0f}%")
````

### 🧐 Lo que acabas de ver

```
🌟 Estudiantes con promedio superior a 4.0:
   - Ana 4.3
   - María 4.8

📋 Lista de destacados: ['Ana', 'María']

📊 Promedio del grupo : 3.85
📊 Aprobados          : 3 de 4
📊 Tasa de aprobación : 75%
```

**Memoriza este patrón de cuatro pasos**, porque lo usarás en cada proyecto de datos de aquí en adelante:

```python
resultado = []                      # 1. Preparo un contenedor vacío
for elemento in coleccion:          # 2. Recorro todo
    if condicion:                   # 3. Decido si me sirve
        resultado.append(elemento)  # 4. Lo guardo
```

> 🧠 **Detalle crítico:** la línea `resultado = []` va **fuera y antes** del ciclo. Si la pones adentro, se vaciaría en cada vuelta y terminarías con un solo elemento. Es un error clásico: si tu lista final tiene solo 1 elemento, revisa dónde la inicializaste.

### 🎓 4.6 · `while`: repetir mientras algo sea cierto

El `for` sirve cuando sabes **qué recorrer**. Pero a veces no sabes cuántas vueltas harán falta: *"sigue ahorrando **hasta** juntar el millón"*, *"sigue creciendo **hasta** superar los 5.000 usuarios"*.

Ahí entra `while`.

```python
while  ahorro < meta :
│      └────┬─────┘ └── dos puntos
│           └── LA CONDICIÓN: mientras esto sea True, se repite
└── palabra clave

    ahorro = ahorro + aporte     ← el cuerpo DEBE acercar la condición a False
```

<div style="background:#FDEDE3;border-radius:10px;padding:16px 20px;margin:14px 0;font-family:Corbel,Calibri,sans-serif">
<b style="color:#C0392B">⚠️ EL BUCLE INFINITO</b><br><br>
Si dentro del <code>while</code> <b>nunca cambias</b> lo que la condición evalúa, el programa se repite <b>para siempre</b>. En Colab lo notarás porque la celda se queda con el círculo girando eternamente.<br><br>
<b>Cómo salir:</b> presiona el botón ⏹️ (detener) al lado de la celda.<br>
<b>Cómo prevenirlo:</b> antes de ejecutar un <code>while</code>, pregúntate <i>"¿qué línea de aquí adentro hará que algún día la condición sea falsa?"</i>. Si no la encuentras, tienes un bucle infinito.
</div>

````python
# ▶️ EJECUTA ESTA CELDA

def simular_ahorro(meta, aporte_mensual):
    """Calcula cuántos meses toma alcanzar una meta de ahorro.

    Parámetros:
        meta (int): cantidad que se quiere ahorrar
        aporte_mensual (int): cuánto se ahorra cada mes

    Devuelve:
        (int, int): los meses necesarios y el total acumulado
    """
    ahorro = 0
    meses = 0

    while ahorro < meta:                    # mientras NO haya llegado a la meta...
        ahorro = ahorro + aporte_mensual    # ...ahorra un mes más   ← esto acerca el final
        meses = meses + 1                   # ...y cuenta el mes

    return meses, ahorro


meses, total = simular_ahorro(meta=1000000, aporte_mensual=150000)

print(f"Meta: $1.000.000 ahorrando $150.000 al mes")
print(f"Necesitas {meses} meses y acumularás ${total:,}")
````

### 🧐 Lo que acabas de ver

```
Meta: $1.000.000 ahorrando $150.000 al mes
Necesitas 7 meses y acumularás $1,050,000
```

**Por qué 7 y no 6.67:** el ciclo solo se detiene cuando `ahorro` **supera o iguala** la meta. Al sexto mes llevabas $900.000 (todavía menos que el millón), así que dio una vuelta más. Por eso el total quedó en $1.050.000 y no exactamente en el millón.

Eso es una característica, no un error: los meses son enteros, no puedes ahorrar 0.67 meses.

**Y la línea que evita el desastre:** `ahorro = ahorro + aporte_mensual`. Es la que hace que la condición `ahorro < meta` se vuelva falsa algún día. Si `aporte_mensual` fuera `0`, `ahorro` se quedaría en 0 para siempre y tendrías un **bucle infinito**.

> 💡 **`{total:,}`** en el f-string agrega los separadores de miles. Detalle pequeño, gran diferencia al presentar resultados.

### 🎓 5.1 · `break` y `continue`

Un ciclo normalmente recorre **todo**. Pero a veces no quieres eso:

- Si estás **buscando** algo, en cuanto lo encuentres no tiene sentido seguir mirando los demás.
- Si estás **limpiando** datos, quieres ignorar los inválidos y seguir con el resto.

Para eso existen dos palabras clave que se escriben **dentro** del ciclo:

| Palabra | Qué hace | Piénsalo como |
|---|---|---|
| **`break`** | **Sale del ciclo** por completo, ahí mismo | *"ya encontré lo que buscaba, me voy"* 🚪 |
| **`continue`** | **Salta** el resto de esta vuelta y pasa a la siguiente | *"este no me sirve, siguiente"* ⏭️ |

```
for elemento in coleccion:
    ...
    if algo:
        break      →  se sale del for por completo
    if otra_cosa:
        continue   →  vuelve arriba con el siguiente elemento
    ...            →  estas líneas NO se ejecutan si hubo continue
```

> 🧹 **En datos:** `break` sirve para **buscar** (paras al encontrar, ahorras tiempo) y `continue` para **filtrar** (te saltas los registros basura sin escribir un `if` gigante que envuelva todo).

````python
# ▶️ EJECUTA ESTA CELDA — break en acción

def buscar_numero(lista, objetivo):
    """Busca un número en una lista y se detiene apenas lo encuentra.

    Parámetros:
        lista (list): lista de números donde buscar
        objetivo (int): el número que se busca
    """
    i = 0
    while i < len(lista):
        print(f"   revisando posición {i}...")
        if lista[i] == objetivo:
            print(f"   ¡Encontrado {objetivo} en la posición {i}!")
            break                    # 🚪 sale del while inmediatamente
        i += 1                       # i += 1 es lo mismo que i = i + 1
    else:
        # Este 'else' pertenece al WHILE (no a un if): se ejecuta solo
        # si el ciclo terminó de forma natural, SIN haber pasado por break.
        print(f"   {objetivo} no se encontró en la lista.")


mis_numeros = [10, 20, 30, 40, 50]

print("Buscando el 30:")
buscar_numero(mis_numeros, 30)

print("\nBuscando el 60:")
buscar_numero(mis_numeros, 60)
````

### 🧐 Lo que acabas de ver

```
Buscando el 30:
   revisando posición 0...
   revisando posición 1...
   revisando posición 2...
   ¡Encontrado 30 en la posición 2!

Buscando el 60:
   revisando posición 0...
   revisando posición 1...
   revisando posición 2...
   revisando posición 3...
   revisando posición 4...
   60 no se encontró en la lista.
```

**Lo importante está en las líneas "revisando":**

- Al buscar el **30**, el ciclo se detuvo en la posición 2. **No revisó las posiciones 3 y 4.** Eso es `break` ahorrándote trabajo: con 5 elementos no se nota, pero con un millón de registros es la diferencia entre segundos y minutos.
- Al buscar el **60**, revisó las 5 posiciones y **nunca ejecutó el `break`**. Por eso se activó el `else` del `while`.

> 🎓 **El `else` de un ciclo** (existe para `for` y `while`) es una rareza de Python que casi nadie conoce: significa *"si el ciclo terminó completo sin interrupciones"*. Es perfecto para el caso "no lo encontré".

````python
# ▶️ EJECUTA ESTA CELDA — continue en acción
# Este es un caso REAL de limpieza de datos: una encuesta con edades mal digitadas.

edades_crudas = ["18", "21", "N/A", "22", "veinte", "20", "", "23"]

edades_limpias = []
descartados = 0

for valor in edades_crudas:
    if not valor.isdigit():        # .isdigit() da True solo si el texto es puro número
        print(f"   ⚠️ descartando: '{valor}'")
        descartados += 1
        continue                   # ⏭️ salta el resto y sigue con el siguiente valor

    # Estas dos líneas solo se ejecutan si el valor SÍ era numérico:
    edades_limpias.append(int(valor))
    print(f"   ✅ aceptando  : {valor}")

print("\n--- RESULTADO ---")
print("Edades válidas   :", edades_limpias)
print("Registros perdidos:", descartados, "de", len(edades_crudas))
print(f"Promedio         : {sum(edades_limpias) / len(edades_limpias):.1f} años")
print(f"Pérdida de datos : {(descartados / len(edades_crudas)) * 100:.1f}%")
````

### 🧐 Lo que acabas de ver

```
   ✅ aceptando  : 18
   ✅ aceptando  : 21
   ⚠️ descartando: 'N/A'
   ✅ aceptando  : 22
   ⚠️ descartando: 'veinte'
   ✅ aceptando  : 20
   ⚠️ descartando: ''
   ✅ aceptando  : 23

--- RESULTADO ---
Edades válidas   : [18, 21, 22, 20, 23]
Registros perdidos: 3 de 8
Promedio         : 20.8 años
Pérdida de datos : 37.5%
```

**Acabas de escribir tu primer proceso de limpieza de datos.** Y contiene dos lecciones profesionales:

1. **`continue` mantiene el código plano.** La alternativa sería envolver todo el cuerpo en un `if valor.isdigit():`, agregando un nivel de sangría. Con muchas reglas de validación, el código se vuelve una escalera ilegible. `continue` lo evita.
2. **Se reporta la pérdida: 37.5 %.** Esto no es un detalle estético. Descartar datos es una **decisión de analista** y debe quedar documentada. Un informe que dice "el promedio de edad es 20.8" sin mencionar que se perdió más de un tercio de las respuestas es un informe incompleto.

> 🧠 Nota que `"veinte"` **sí contiene una edad** que cualquier humano lee. La descartaste porque tu regla era simple. Un proceso más sofisticado la recuperaría. Siempre hay una decisión detrás.

### 🎓 5.2 · Funciones iterativas vs. recursivas

Existen **dos maneras** de resolver un problema repetitivo, y conviene conocer las dos:

| Tipo | Cómo funciona | Ejemplo típico |
|---|---|---|
| **Iterativa** | **No se llama a sí misma**; ejecuta la tarea con un **bucle** | Sumar una lista, recorrer registros |
| **Recursiva** | **Se llama a sí misma** con un problema más pequeño, hasta llegar a un caso trivial | Factorial, recorrer carpetas dentro de carpetas |

> 🪆 **La analogía:** una función recursiva es como una **matrioska** (muñeca rusa). Para abrir la muñeca de tamaño 5, primero abres la de tamaño 4, que necesita abrir la de 3… hasta llegar a la más pequeña, que ya no tiene nada dentro. Esa última es el **caso base**.

Toda función recursiva necesita **exactamente dos cosas**:

| Ingrediente | Qué es | Si falta… |
|---|---|---|
| **1. Caso base** | La condición que **detiene** la recursión | `RecursionError`: se llama infinitamente hasta agotar la memoria |
| **2. Paso recursivo** | La llamada a sí misma con un problema **más pequeño** | Nunca llega al caso base → mismo error |

**El factorial** es el ejemplo clásico: `5! = 5 × 4 × 3 × 2 × 1 = 120`. Y se puede definir de forma recursiva porque `5! = 5 × 4!`

````python
# ▶️ EJECUTA ESTA CELDA
# El mismo cálculo, resuelto de las dos formas

def factorial_iterativo(n):
    """Calcula n! usando un bucle for."""
    resultado = 1
    for i in range(1, n + 1):
        resultado = resultado * i
    return resultado


def factorial_recursivo(n):
    """Calcula n! llamándose a sí misma."""
    if n <= 1:                                   # 1️⃣ CASO BASE: detiene la recursión
        return 1
    else:
        return n * factorial_recursivo(n - 1)    # 2️⃣ PASO RECURSIVO: n-1 es más pequeño


print("Factorial de 5 (iterativo):", factorial_iterativo(5))
print("Factorial de 5 (recursivo):", factorial_recursivo(5))
print("Factorial de 0 (recursivo):", factorial_recursivo(0))
````

### 🧐 Lo que acabas de ver

```
Factorial de 5 (iterativo): 120
Factorial de 5 (recursivo): 120
Factorial de 0 (recursivo): 1
```

Las dos versiones dan lo mismo. Pero **cómo llegan** es muy distinto.

**Así se desarma la recursión** cuando llamas `factorial_recursivo(5)`:

```
factorial(5) = 5 * factorial(4)          ← se queda esperando...
             = 5 * 4 * factorial(3)      ← ...y esta también...
             = 5 * 4 * 3 * factorial(2)  ← ...y esta...
             = 5 * 4 * 3 * 2 * factorial(1)
             = 5 * 4 * 3 * 2 * 1         ← ¡caso base! devuelve 1
             = 120                       ← ahora sí se resuelve todo hacia atrás
```

Python va **apilando** llamadas pendientes hasta llegar al caso base, y después las resuelve en orden inverso.

### ⚠️ Qué pasa si olvidas el caso base

Si quitaras el `if n <= 1: return 1`, la función se llamaría con 4, 3, 2, 1, 0, -1, -2… **para siempre**. Python te protege deteniéndolo con:

```
RecursionError: maximum recursion depth exceeded
```

> 🤔 **¿Cuál usar entonces?** En ciencia de datos, casi siempre la **iterativa**: es más fácil de leer y no tiene límite de profundidad. La recursión brilla en estructuras anidadas (carpetas dentro de carpetas, categorías dentro de categorías). Conviene conocerla, pero no forzarla.

### 🎓 6.4 · NumPy: el motor numérico

En la clase pasada descubriste algo raro: si `notas` es una lista y escribes `notas * 2`, Python **duplica la lista** en vez de multiplicar los valores.

NumPy resuelve exactamente eso. Su estructura estrella es el **`array`**: parecido a una lista, pero con dos diferencias enormes:

| | Lista de Python | Array de NumPy |
|---|---|---|
| Tipos | Puede mezclar texto, números, lo que sea | **Un solo tipo** para todos los elementos |
| Operaciones | `lista * 2` duplica la lista | `array * 2` **multiplica cada valor** |
| Velocidad | Lenta para cálculo masivo | **Hasta 50 veces más rápida** |

A esa capacidad de operar sobre todos los elementos a la vez, sin escribir un `for`, se le llama **vectorización**. Es la razón por la que existe NumPy.

> ⚡ **Por qué importa:** con 5 datos no notarás diferencia. Con 5 millones —el tamaño normal de un dataset real— un `for` en Python puro puede tardar minutos y NumPy, milésimas de segundo.

````python
# ▶️ EJECUTA ESTA CELDA
import numpy as np       # el alias estándar

# Primero, el problema con las listas normales:
notas_lista = [4.0, 3.5, 2.8, 4.7, 3.1]
print("Con una LISTA, notas * 2 da:")
print("  ", notas_lista * 2, "  😱 duplicó la lista")

# Ahora con NumPy:
notas = np.array([4.0, 3.5, 2.8, 4.7, 3.1])

print("\nCon un ARRAY de NumPy:")
print("   Notas originales :", notas)
print("   notas * 2        :", notas * 2, "  ✅ multiplicó cada valor")
print("   notas + 0.5      :", notas + 0.5, "  (medio punto para todos)")
print("   notas * 20       :", notas * 20, "  (convertidas a escala 100)")

print("\n📊 Estadísticos en una línea cada uno:")
print("   Promedio            :", notas.mean())
print("   Máxima / mínima     :", notas.max(), "/", notas.min())
print("   Desviación estándar :", round(notas.std(), 3))
print("   Suma total          :", notas.sum())

print("\n🎯 Filtrado vectorizado (¡sin escribir ningún for!):")
print("   Notas aprobadas :", notas[notas >= 3.0])
print("   ¿Cuáles aprueban?:", notas >= 3.0)
````

### 🧐 Lo que acabas de ver

```
Con una LISTA, notas * 2 da:
   [4.0, 3.5, 2.8, 4.7, 3.1, 4.0, 3.5, 2.8, 4.7, 3.1]   😱 duplicó la lista

Con un ARRAY de NumPy:
   Notas originales : [4.  3.5 2.8 4.7 3.1]
   notas * 2        : [8.  7.  5.6 9.4 6.2]   ✅ multiplicó cada valor
   ...

🎯 Filtrado vectorizado (¡sin escribir ningún for!):
   Notas aprobadas : [4.  3.5 4.7 3.1]
   ¿Cuáles aprueban?: [ True  True False  True  True]
```

**Las dos ideas más importantes de NumPy están en esas últimas líneas:**

1. **`notas >= 3.0` no devuelve un solo `True`**, sino **una lista de `True`/`False`, uno por cada nota**. Se llama *máscara booleana*.
2. **`notas[notas >= 3.0]`** usa esa máscara como filtro: *"dame solo los elementos donde la máscara dice `True`"*.

Compáralo con lo que habrías escrito en la Parte 4:

```python
# Antes (Python puro): 4 líneas
aprobadas = []
for nota in notas:
    if nota >= 3.0:
        aprobadas.append(nota)

# Ahora (NumPy): 1 línea
aprobadas = notas[notas >= 3.0]
```

> 🔑 Esta sintaxis de filtrado es **exactamente la misma que usarás en pandas** para filtrar tablas enteras. Vale la pena entenderla bien aquí, con 5 números, antes de aplicarla a 50.000 filas.

```python
def auditar_datos(df):
    """Audita la calidad de un DataFrame y entrega un puntaje de 0 a 100.

    Parámetros:
        df (DataFrame): la tabla a auditar

    Devuelve:
        int: el puntaje de calidad (100 = sin problemas detectados)
    """
    problemas = 0

    print("=" * 58)
    print("            AUDITORÍA DE CALIDAD DE DATOS")
    print("=" * 58)

    # 1) Dimensiones
    filas, columnas = df.shape
    print(f"Dimensiones: {filas} filas x {columnas} columnas\n")

    # 2) Recorrer cada columna
    print("COLUMNAS:")
    for col in df.columns:
        nulos = df[col].isnull().sum()
        marca = "⚠️" if nulos > 0 else "  "
        print(f"  {marca} {col:<12} tipo: {str(df[col].dtype):<10} nulos: {nulos}")
        if nulos > 0:
            problemas += 1

    # 3) Duplicados
    duplicados = df.duplicated().sum()
    print(f"\nFilas duplicadas: {duplicados}")
    if duplicados > 0:
        problemas += 1

    # 4) Valores únicos en 'ciudad'
    if "ciudad" in df.columns:
        unicos = df["ciudad"].nunique()
        print(f"Ciudades distintas según la computadora: {unicos}")
        print(f"  -> {list(df['ciudad'].dropna().unique())}")
        print("  ⚠️ Revisa si son ciudades reales o errores de formato")
        problemas += 1

    # 5) Problemas en columnas numéricas
    print("\nVALIDACIÓN DE RANGOS:")
    if "edad" in df.columns:
        edades_malas = df[(df["edad"] < 0) | (df["edad"] > 120)]
        if len(edades_malas) > 0:
            print(f"  ⚠️ {len(edades_malas)} edad(es) fuera del rango 0-120")
            problemas += 1

    if "ingresos" in df.columns:
        ingresos_malos = df[df["ingresos"] < 0]
        if len(ingresos_malos) > 0:
            print(f"  ⚠️ {len(ingresos_malos)} ingreso(s) negativo(s)")
            problemas += 1

    # 6) Puntaje final
    puntaje = 100 - (problemas * 10)
    if puntaje < 0:
        puntaje = 0

    print("\n" + "=" * 58)
    print(f"  PROBLEMAS DETECTADOS: {problemas}")
    print(f"  PUNTAJE DE CALIDAD  : {puntaje}/100")
    print("=" * 58)

    return puntaje
```

````python
# 🥇 RETO C — Tus datos y tus funciones
# Mi tema es: mis horas de sueño de los últimos días

registros_sueno = [
    {"dia": "2026-07-20", "horas_sueno": 7.5, "seguimiento_app": True},
    {"dia": "2026-07-21", "horas_sueno": 6.0, "seguimiento_app": True},
    {"dia": "2026-07-22", "horas_sueno": 5.5, "seguimiento_app": False},
    {"dia": "2026-07-23", "horas_sueno": 8.0, "seguimiento_app": True},
    {"dia": "2026-07-24", "horas_sueno": 7.0, "seguimiento_app": True},
    {"dia": "2026-07-25", "horas_sueno": 4.5, "seguimiento_app": False},
    {"dia": "2026-07-26", "horas_sueno": 9.0, "seguimiento_app": True},
    {"dia": "2026-07-27", "horas_sueno": 7.2, "seguimiento_app": True},
    {"dia": "2026-07-28", "horas_sueno": 6.5, "seguimiento_app": True},
    {"dia": "2026-07-29", "horas_sueno": 8.5, "seguimiento_app": True},
    {"dia": "2026-07-30", "horas_sueno": 5.0, "seguimiento_app": False},
    {"dia": "2026-07-31", "horas_sueno": 7.8, "seguimiento_app": True},
]


def clasificar_sueno(horas):
    """Clasifica una noche de sueño en óptimo, suficiente o insuficiente.

    Óptimo: entre 7 y 9 horas · Suficiente: entre 6 y 7 · Insuficiente: lo demás.
    """
    if horas >= 7 and horas <= 9:
        return "óptimo"
    elif horas >= 6:
        return "suficiente"
    else:
        return "insuficiente"


def promedio_horas(registros):
    """Calcula el promedio de horas de sueño de una lista de registros."""
    total = 0
    for registro in registros:
        total = total + registro["horas_sueno"]
    return total / len(registros)


def contar_categoria(registros, categoria):
    """Cuenta cuántos registros caen en una categoría de sueño dada."""
    contador = 0
    for registro in registros:
        if clasificar_sueno(registro["horas_sueno"]) == categoria:
            contador = contador + 1
    return contador


def filtrar_por_seguimiento(registros, con_app=True):
    """Filtra los registros según si se registraron con la app (por defecto True)."""
    seleccion = []
    for registro in registros:
        if registro["seguimiento_app"] == con_app:
            seleccion.append(registro)
    return seleccion
````

````python
# 🥇 RETO C — Tu informe final

import numpy as np

horas = np.array([r["horas_sueno"] for r in registros_sueno])

print("=" * 50)
print("         INFORME: MIS HORAS DE SUEÑO")
print("=" * 50)
print(f"Días registrados : {len(registros_sueno)}")
print(f"Promedio         : {promedio_horas(registros_sueno):.2f} h")
print(f"Mínima / máxima  : {horas.min()} / {horas.max()} h")
print(f"Desviación est.  : {horas.std():.2f} h")

print("\nClasificación:")
for categoria in ["óptimo", "suficiente", "insuficiente"]:
    n = contar_categoria(registros_sueno, categoria)
    print(f"  {categoria:<12}: {n} noche(s) ({(n / len(registros_sueno)) * 100:.0f}%)")

print("\nSegún cómo se registró:")
con_app = filtrar_por_seguimiento(registros_sueno)
sin_app = filtrar_por_seguimiento(registros_sueno, con_app=False)
print(f"  Con app: {len(con_app)} registros, promedio {promedio_horas(con_app):.2f} h")
print(f"  Sin app: {len(sin_app)} registros, promedio {promedio_horas(sin_app):.2f} h")
print("=" * 50)
````
