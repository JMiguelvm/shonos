## 1.3 Herramientas de carga y auditoría de calidad

La auditoría responde **antes** de limpiar: ¿cuántas filas, cuántos faltantes, cuántos
duplicados, cuántos atípicos IQR tiene cada columna? Esto define qué hay que corregir.

```python
def cargar_datos(ruta, **kwargs):
    """Carga un .csv y devuelve un DataFrame; si falla, avisa y devuelve None."""
    extension = str(ruta).split(".")[-1].lower()
    try:
        if extension == "csv":
            df = pd.read_csv(ruta, **kwargs)
        else:
            print(f"Extensión '.{extension}' no reconocida. Usa .csv.")
            return None
    except FileNotFoundError:
        print(f"No encuentro el archivo '{ruta}'. Revisa el nombre.")
        return None
    except Exception as error:
        print(f"El archivo existe pero no se pudo leer. Detalle: {error}")
        return None
    print(f"Cargado {Path(ruta).name}: {df.shape[0]} filas, {df.shape[1]} columnas.")
    return df


def detectar_atipicos_iqr(serie):
    """True donde el valor es atípico según la regla de 1.5×IQR."""
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1
    return (serie < q1 - 1.5 * iqr) | (serie > q3 + 1.5 * iqr)


def auditoria_calidad(df):
    """Reporte: tipo, % faltantes, únicos, atípicos y diagnóstico por columna."""
    filas_reporte = []
    for columna in df.columns:
        pct_nulos = df[columna].isna().mean() * 100
        n_unicos = df[columna].nunique()
        if pct_nulos > 40:
            estado = "Inutilizable"
        elif pct_nulos > 10:
            estado = "Requiere tratamiento"
        else:
            estado = "Aceptable"
        if str(df[columna].dtype) in ("object", "string") and n_unicos > len(df) * 0.5:
            estado += " · posible identificador"
        if pd.api.types.is_numeric_dtype(df[columna]) and not pd.api.types.is_bool_dtype(df[columna]):
            n_atipicos = detectar_atipicos_iqr(df[columna].dropna()).sum()
        else:
            n_atipicos = "-"
        filas_reporte.append({
            "columna": columna, "tipo": str(df[columna].dtype),
            "% faltantes": round(pct_nulos, 1), "valores_unicos": n_unicos,
            "atipicos_iqr": n_atipicos, "diagnostico": estado,
        })
    reporte = pd.DataFrame(filas_reporte)
    print(f"Puntaje general de calidad: {round(100 - reporte['% faltantes'].mean(), 1)}/100")
    print(f"Filas totales: {len(df)}  ·  Filas duplicadas: {df.duplicated().sum()}")
    return reporte
```

```python
# -*- coding: utf-8 -*-
"""EDA + análisis estadístico + hallazgos sobre siniestros_viales_limpio.csv"""

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, kruskal

sns.set_style("whitegrid")
FIGDIR = "/"

df = pd.read_csv("siniestros_viales_limpio.csv", sep=";", low_memory=False)
print("Filas:", df.shape[0], "Columnas:", df.shape[1])

TOTAL_FALLECIDOS = df["conteoregistros"].sum()
print(f"Total de personas fallecidas (ponderado): {int(TOTAL_FALLECIDOS):,}")

# ------------------------------------------------------------------
# Helpers
# ------------------------------------------------------------------
def distribucion_ponderada(df, columna, top=15):
    d = df.dropna(subset=[columna])
    tabla = d.groupby(columna)["conteoregistros"].sum().sort_values(ascending=False)
    pct = (tabla / tabla.sum() * 100).round(2)
    return pd.DataFrame({"fallecidos": tabla, "%": pct}).head(top)

def cramers_v(df, col1, col2, pesos="conteoregistros"):
    d = df.dropna(subset=[col1, col2])
    tabla = pd.pivot_table(d, values=pesos, index=col1, columns=col2, aggfunc="sum", fill_value=0)
    chi2, p, gl, _ = chi2_contingency(tabla)
    n = tabla.values.sum()
    k = min(tabla.shape) - 1
    v = np.sqrt(chi2 / (n * k)) if k > 0 else np.nan
    return chi2, p, gl, v

def prueba_chi2(df, col1, col2, alfa=0.05):
    chi2, p, gl, v = cramers_v(df, col1, col2)
    concl = "SÍ hay asociación (se rechaza H0)" if p < alfa else "sin evidencia de asociación"
    print(f"  {col1} vs {col2}: chi2={chi2:.1f}  gl={gl}  p={p:.4g}  V Cramér={v:.3f}  -> {concl}")
    return chi2, p, v

# ==================================================================
# 1. EDA - Distribuciones univariantes
# ==================================================================
print("\n" + "="*60 + "\n1. EDA - DISTRIBUCIONES UNIVARIANTES\n" + "="*60)

zona_dist = distribucion_ponderada(df, "zona")
sexo_dist = distribucion_ponderada(df, "sexo")
usuario_dist = distribucion_ponderada(df, "usuariovia")
vehiculo_dist = distribucion_ponderada(df, "tipovehiculogrupo")
depto_dist = distribucion_ponderada(df, "departamento", top=10)
anio_dist = df.groupby("anohecho")["conteoregistros"].sum().sort_index()
dia_dist = df.dropna(subset=["diaocurrencia"]).groupby("diaocurrencia")["conteoregistros"].sum()
franja_dist = df.dropna(subset=["rango3horas"]).groupby("rango3horas")["conteoregistros"].sum()

print("\nZona:\n", zona_dist)
print("\nSexo:\n", sexo_dist)
print("\nUsuario de la vía:\n", usuario_dist)
print("\nTipo de vehículo:\n", vehiculo_dist)
print("\nTop 10 departamentos:\n", depto_dist)
print("\nPor año:\n", anio_dist)

# Edad ponderada (expandida por ConteoRegistros)
df_edad = df.dropna(subset=["edad"]).copy()
edades_ponderadas = np.repeat(df_edad["edad"].values, df_edad["conteoregistros"].astype(int).values)
print("\nEstadísticos de edad (ponderados):\n", pd.Series(edades_ponderadas).describe())

# --- Figura 1: panel univariante ---
fig, axes = plt.subplots(2, 3, figsize=(19, 10))

zona_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[0,0], color="#003366")
axes[0,0].set_title("Fallecidos por zona"); axes[0,0].set_xlabel("Fallecidos")

sexo_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[0,1], color="#0059b3")
axes[0,1].set_title("Fallecidos por sexo"); axes[0,1].set_xlabel("Fallecidos")

sns.histplot(edades_ponderadas, bins=30, kde=True, color="#003366", ax=axes[0,2])
axes[0,2].axvline(np.median(edades_ponderadas), color="red", linestyle="--", label="Mediana")
axes[0,2].set_title("Distribución de edad"); axes[0,2].legend()

usuario_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,0], color="#336699")
axes[1,0].set_title("Fallecidos por usuario de la vía"); axes[1,0].set_xlabel("Fallecidos")

vehiculo_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,1], color="#6699cc")
axes[1,1].set_title("Fallecidos por tipo de vehículo"); axes[1,1].set_xlabel("Fallecidos")

depto_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,2], color="#00264d")
axes[1,2].set_title("Top 10 departamentos"); axes[1,2].set_xlabel("Fallecidos")

plt.tight_layout()
plt.savefig(f"{FIGDIR}/01_distribuciones_univariantes.png", dpi=130)
plt.close()

# --- Figura 2: evolución temporal ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
anio_dist.plot(kind="line", marker="o", ax=axes[0], color="#003366")
axes[0].set_title("Evolución anual de fallecidos"); axes[0].set_xlabel("Año"); axes[0].set_ylabel("Fallecidos")

dia_dist.reindex(sorted(dia_dist.index)).plot(kind="bar", ax=axes[1], color="#336699")
axes[1].set_title("Fallecidos por día de la semana"); axes[1].tick_params(axis='x', rotation=45)

franja_dist.plot(kind="bar", ax=axes[2], color="#6699cc")
axes[2].set_title("Fallecidos por franja horaria"); axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig(f"{FIGDIR}/02_evolucion_temporal.png", dpi=130)
plt.close()

# ==================================================================
# 2. EDA - Relaciones bivariantes
# ==================================================================
print("\n" + "="*60 + "\n2. EDA - RELACIONES BIVARIANTES\n" + "="*60)

def tabla_cruzada_pct(df, fila, columna):
    d = df.dropna(subset=[fila, columna])
    tabla = pd.pivot_table(d, values="conteoregistros", index=fila, columns=columna, aggfunc="sum", fill_value=0)
    return tabla.div(tabla.sum(axis=1), axis=0).mul(100).round(1)

zona_usuario = tabla_cruzada_pct(df, "zona", "usuariovia")
sexo_usuario = tabla_cruzada_pct(df, "sexo", "usuariovia")
print("\nZona x Usuario de la vía (% fila):\n", zona_usuario)
print("\nSexo x Usuario de la vía (% fila):\n", sexo_usuario)

df_pond = df.dropna(subset=["edad", "zona", "sexo"]).copy()
df_pond = df_pond.loc[df_pond.index.repeat(df_pond["conteoregistros"].astype(int))]

dia_franja = None
if {"diaocurrencia", "rango3horas"}.issubset(df.columns):
    d = df.dropna(subset=["diaocurrencia", "rango3horas"])
    dia_franja = pd.pivot_table(d, values="conteoregistros", index="diaocurrencia",
                                 columns="rango3horas", aggfunc="sum", fill_value=0)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

sns.heatmap(zona_usuario, annot=True, fmt=".1f", cmap="Blues", ax=axes[0,0])
axes[0,0].set_title("% usuario de la vía dentro de cada zona")

sns.boxplot(data=df_pond, x="zona", y="edad", palette="Blues", ax=axes[0,1])
axes[0,1].set_title("Edad por zona")

sns.boxplot(data=df_pond, x="sexo", y="edad", palette="Blues", ax=axes[1,0])
axes[1,0].set_title("Edad por sexo")

if dia_franja is not None:
    dia_franja_plot = dia_franja.drop(index="Sin información", columns="Sin información", errors="ignore")
    sns.heatmap(dia_franja_plot.reindex(sorted(dia_franja_plot.index)), cmap="Reds", ax=axes[1,1])
    axes[1,1].set_title("Fallecidos: día de la semana x franja horaria\n(excluye registros 'Sin información')")

plt.tight_layout()
plt.savefig(f"{FIGDIR}/03_relaciones_bivariantes.png", dpi=130)
plt.close()

# ==================================================================
# 3. Fuerza de asociación (V de Cramér) entre categóricas
# ==================================================================
print("\n" + "="*60 + "\n3. V DE CRAMÉR ENTRE VARIABLES CATEGÓRICAS\n" + "="*60)

VARS_CAT = [c for c in ["zona", "sexo", "usuariovia", "tipovehiculogrupo",
                         "objetocolision", "condicionvictima", "departamento",
                         "diaocurrencia", "rango3horas"] if c in df.columns]

matriz_v = pd.DataFrame(index=VARS_CAT, columns=VARS_CAT, dtype=float)
for a in VARS_CAT:
    for b in VARS_CAT:
        if a == b:
            matriz_v.loc[a, b] = 1.0
        else:
            try:
                _, _, _, v = cramers_v(df, a, b)
                matriz_v.loc[a, b] = v
            except Exception:
                matriz_v.loc[a, b] = np.nan

print(matriz_v.round(3))

plt.figure(figsize=(9, 7))
sns.heatmap(matriz_v.astype(float), annot=True, fmt=".2f", cmap="viridis", vmin=0, vmax=1)
plt.title("V de Cramér entre variables categóricas")
plt.tight_layout()
plt.savefig(f"{FIGDIR}/04_cramers_v.png", dpi=130)
plt.close()

# ==================================================================
# 4. Análisis estadístico - Pruebas de hipótesis (alfa = 0.05)
# ==================================================================
print("\n" + "="*60 + "\n4. PRUEBAS DE HIPÓTESIS (alfa = 0.05)\n" + "="*60)

print("\nChi-cuadrado de independencia:")
r_zona_usuario = prueba_chi2(df, "zona", "usuariovia")
r_zona_vehiculo = prueba_chi2(df, "zona", "tipovehiculogrupo")
r_sexo_usuario = prueba_chi2(df, "sexo", "usuariovia")
r_zona_sexo = prueba_chi2(df, "zona", "sexo")
if "diaocurrencia" in df and "rango3horas" in df:
    r_dia_franja = prueba_chi2(df, "diaocurrencia", "rango3horas")

print("\nKruskal-Wallis (¿difiere la edad entre grupos?):")
grupos_zona = [df_pond.loc[df_pond["zona"] == z, "edad"] for z in df_pond["zona"].unique()]
h_zona, p_zona = kruskal(*grupos_zona)
print(f"  Edad según zona: H={h_zona:.1f}  p={p_zona:.4g}  -> "
      f"{'diferencia significativa' if p_zona < 0.05 else 'sin diferencia significativa'}")

grupos_sexo = [df_pond.loc[df_pond["sexo"] == s, "edad"] for s in df_pond["sexo"].unique()]
h_sexo, p_sexo = kruskal(*grupos_sexo)
print(f"  Edad según sexo: H={h_sexo:.1f}  p={p_sexo:.4g}  -> "
      f"{'diferencia significativa' if p_sexo < 0.05 else 'sin diferencia significativa'}")

df_pond_u = df.dropna(subset=["edad", "usuariovia"]).copy()
df_pond_u = df_pond_u.loc[df_pond_u.index.repeat(df_pond_u["conteoregistros"].astype(int))]
grupos_usuario = [df_pond_u.loc[df_pond_u["usuariovia"] == u, "edad"] for u in df_pond_u["usuariovia"].unique()]
h_usr, p_usr = kruskal(*grupos_usuario)
print(f"  Edad según usuario de la vía: H={h_usr:.1f}  p={p_usr:.4g}  -> "
      f"{'diferencia significativa' if p_usr < 0.05 else 'sin diferencia significativa'}")

# ==================================================================
# 5. Hallazgos, patrones e insights (generados desde los datos reales)
# ==================================================================
print("\n" + "="*60 + "\n5. HALLAZGOS\n" + "="*60)

top_departamento = depto_dist.index[0]; pct_top_departamento = depto_dist["%"].iloc[0]
zona_mayoritaria = zona_dist.index[0]; pct_zona_mayoritaria = zona_dist["%"].iloc[0]
usuario_mayoritario = usuario_dist.index[0]; pct_usuario_mayoritario = usuario_dist["%"].iloc[0]
sexo_mayoritario = sexo_dist.index[0]; pct_sexo_mayoritario = sexo_dist["%"].iloc[0]
vehiculo_mayoritario = vehiculo_dist.index[0]; pct_vehiculo_mayoritario = vehiculo_dist["%"].iloc[0]
edad_mediana = np.median(edades_ponderadas)
anio_pico = anio_dist.idxmax(); fallecidos_anio_pico = int(anio_dist.max())
dia_pico = dia_dist.drop(labels=["Sin información"], errors="ignore").idxmax() if not dia_dist.empty else None
franja_pico = franja_dist.drop(labels=["Sin información"], errors="ignore").idxmax() if not franja_dist.empty else None
zona_usuario_top_urbana = zona_usuario.loc["Urbana"].idxmax() if "Urbana" in zona_usuario.index else None
zona_usuario_top_rural = zona_usuario.loc["Rural"].idxmax() if "Rural" in zona_usuario.index else None

hallazgos = f"""
1. CONCENTRACIÓN GEOGRÁFICA: {top_departamento} concentra el {pct_top_departamento}% de los
   fallecidos del dataset ({depto_dist['fallecidos'].iloc[0]:,.0f} de {TOTAL_FALLECIDOS:,.0f}).

2. ZONA: el {pct_zona_mayoritaria}% de los fallecidos ocurre en zona {str(zona_mayoritaria).lower()}.
   Chi-cuadrado zona-usuariovia: p={r_zona_usuario[1]:.4g}, V Cramér={r_zona_usuario[2]:.3f}
   ({'asociación' if r_zona_usuario[1] < 0.05 else 'sin asociación'} estadísticamente significativa).
   En zona urbana el perfil dominante es "{zona_usuario_top_urbana}"; en zona rural, "{zona_usuario_top_rural}".

3. ACTOR VIAL MÁS VULNERABLE: "{usuario_mayoritario}" representa el {pct_usuario_mayoritario}%
   de los fallecidos, el grupo de mayor riesgo. El vehículo más involucrado es
   "{vehiculo_mayoritario}" ({pct_vehiculo_mayoritario}%).

4. SEXO: el {pct_sexo_mayoritario}% de los fallecidos es "{sexo_mayoritario}".
   Kruskal-Wallis edad~sexo: H={h_sexo:.1f}, p={p_sexo:.4g}
   ({'diferencia' if p_sexo < 0.05 else 'sin diferencia'} significativa de edad entre sexos).

5. EDAD: mediana de {edad_mediana:.0f} años. Kruskal-Wallis edad~zona: p={p_zona:.4g};
   edad~usuario de la vía: p={p_usr:.4g}.

6. TENDENCIA TEMPORAL: {int(anio_pico)} es el año con más fallecidos ({fallecidos_anio_pico:,}).
   Día con más fallecidos: {dia_pico}. Franja horaria más crítica (excluyendo "Sin información",
   que representa el {franja_dist.get('Sin información', 0) / franja_dist.sum() * 100:.1f}% de
   los registros con franja): {franja_pico}.
"""
print(hallazgos)

with open(f"{FIGDIR}/hallazgos.txt", "w", encoding="utf-8") as f:
    f.write(hallazgos)

print("\nListo. Figuras guardadas en", FIGDIR)
```

## 1.5 Funciones de normalización (Paso 3 y Paso 4)

- `normalizar_clave`: minúsculas, sin tildes, espacios colapsados → clave comparable.
- `limpiar_numerico`: "1,0" / " 1.0 " → 1.0 (respeta la coma decimal).
- `limpiar_fecha`: str → `datetime`, lo no parseable → NaN.
- Categorías con tipeos se resuelven con **similitud difusa** (`SequenceMatcher`, umbral 0.6).

```python
def normalizar_clave(texto):
    """Clave estándar para comparar categorías (minúsculas, sin tildes)."""
    if pd.isna(texto):
        return None
    texto = str(texto).strip().lower()
    for a, b in [("á", "a"), ("é", "e"), ("í", "i"), ("ó", "o"),
                 ("ú", "u"), ("ü", "u"), ("ñ", "n")]:
        texto = texto.replace(a, b)
    return " ".join(texto.split())


def limpiar_numerico(serie):
    """'1,0' / ' 1.0 ' / '23.0' -> float. Respeta la coma decimal."""
    limpio = serie.astype("string").str.strip().str.replace(",", ".", regex=False)
    return pd.to_numeric(limpio, errors="coerce").astype(float)


def limpiar_fecha(serie):
    """Convierte a fecha real; lo que no se puede parsear queda NaN."""
    return pd.to_datetime(serie, errors="coerce")


def resolver_por_similitud(clave, canonicos, umbral=0.6):
    """Categoría canónica (con formato) más parecida a `clave`."""
    canonicos_clave = [normalizar_clave(c) for c in canonicos]
    mejor, razon = None, umbral
    for i, can in enumerate(canonicos_clave):
        r = SequenceMatcher(None, clave, can).ratio()
        if r >= razon:
            mejor, razon = canonicos[i], r
    return mejor


def mejor_categoria(clave, canonicos, caso_especial=None):
    """Clave -> canónica: caso especial (p.ej. 'Na' = 'No Aplica'), si no difusa."""
    if caso_especial and clave in caso_especial:
        return caso_especial[clave]
    return resolver_por_similitud(clave, canonicos)


def limpiar_dia(serie):
    """'7.domingo', '6.SÁBADO', '  1' -> '6.Sábado' (canónico)."""
    num = pd.to_numeric(serie.astype("string").str.extract(r"^\s*(\d+)")[0], errors="coerce")
    return pd.Series([etiqueta_dia(n) for n in num], index=serie.index)


def limpiar_mes(serie):
    """'01.ENERO', '03.marzo', '01Enero' -> '03.Marzo' (canónico)."""
    num = pd.to_numeric(serie.astype("string").str.extract(r"^\s*(\d+)")[0], errors="coerce")
    return pd.Series([f"{int(n):02d}.{MESES[int(n)]}" if pd.notna(n) and int(n) in MESES else np.nan
                      for n in num], index=serie.index)


def limpiar_rango(serie):
    """'18:00 A 20:59', ' 15:00  a  17:59' -> '18:00 a 20:59' (canónico)."""
    partes = serie.astype("string").str.extract(r"(\d{2}):(\d{2})\s*[aA]\s*(\d{2}):(\d{2})")
    resultado = []
    for fila in partes.itertuples(index=False):
        if pd.isna(fila[0]):
            resultado.append(np.nan)
        else:
            resultado.append(f"{int(fila[0]):02d}:{int(fila[1]):02d} a "
                             f"{int(fila[2]):02d}:{int(fila[3]):02d}")
    return pd.Series(resultado, index=serie.index)


CANONICOS_USUARIO_VIA = [
    "Usuario de moto", "Peatón", "Usuario de V.Individual", "Usuario de bicicleta",
    "Usuario T.Carga", "Usuario T.Pasajeros", "Usuario de otros", "Sin información",
]
CANONICOS_ACTOR_VIAL = [
    "Usuario de moto", "Peatón", "Usuario de vehículo", "Usuario de bicicleta",
    "Usuario otros", "Sin información",
]
CANONICOS_SEXO = ["Hombre", "Mujer", "Indeterminado"]
CANONICOS_ZONA = ["Urbana", "Rural", "Sin información"]
CANONICOS_TIPO_VEHICULO = [
    "Motocicleta", "No Aplica", "Transporte Individual", "Bicicleta",
    "Transporte De Carga", "Transporte De Pasajeros", "Otras Motos",
    "Maq. Agrícola", "Maq. Industrial", "Otros", "Movilidad Asistida",
]
CANONICOS_OBJETO_COLISION = [
    "Motocicleta", "Objeto fijo", "Transporte individual", "Transporte de carga",
    "No aplica", "Transporte de pasajeros", "Sin información", "Animal en vía",
    "Bicicleta", "Otros", "Maq. industrial", "Maq. agrícola",
]
CANONICOS_CONDICION = ["Conductor", "Peatón", "Pasajero", "Sin información"]
CASO_ESPECIAL_NA = {"na": "No Aplica"}


def aplicar_categoria(serie, canonicos, caso_especial=None):
    """Resuelve cada texto a su categoría canónica (o NaN si no coincide nada)."""
    mapa = {}
    for clave in serie.map(normalizar_clave).dropna().unique():
        mapa[clave] = mejor_categoria(clave, canonicos, caso_especial)
    return serie.map(normalizar_clave).map(
        lambda k: mapa.get(k) if isinstance(k, str) else np.nan)


def limpiar_categoria(serie, canonicos, caso_especial=None, por_defecto="Sin información"):
    """Paso 4: aplica equivalencias y marca lo no resuelto como 'Sin información'."""
    return aplicar_categoria(serie, canonicos, caso_especial).fillna(por_defecto)
```

## 1.6 Georreferencia: el código municipal es la llave canónica

El código DIVIPOLA (5 dígitos) manda sobre el texto. El municipio canónico se obtiene por
**moda** del texto que acompañó a cada código (el texto es el que viene sucio); el
**departamento** se deduce de los 2 primeros dígitos del código.

```python
def limpiar_codigo_municipio(serie):
    """'5001.0', ' 66001,0' -> '05001' (str de 5 dígitos). Si el prefijo no
    corresponde a un departamento válido, queda NaN."""
    num = limpiar_numerico(serie)
    out = num.map(lambda x: f"{int(x):05d}" if pd.notna(x) and int(x) > 0 else np.nan)
    return out.map(lambda c: c if (pd.notna(c) and c[:2] in CODIGO_DEPARTAMENTO) else np.nan)


def estandarizar_nombre_municipio(serie):
    """Texto municipal: Título limpio + diccionario de tildes."""
    limpio = serie.map(normalizar_clave)
    titulo = limpio.map(lambda t: " ".join(w.capitalize() for w in t.split())
                        if isinstance(t, str) else np.nan)
    return titulo.map(lambda t: ACENTOS_MUNICIPIO.get(t, t) if isinstance(t, str) else np.nan)


def construir_mapa_municipio(dfs):
    """Nombre canónico por código municipal (moda del texto en los 3 archivos)."""
    partes = []
    for df in dfs:
        df = df.copy()
        df.columns = df.columns.str.lower()
        cod = limpiar_codigo_municipio(df["codigomunicipiohecho"])
        nombre = estandarizar_nombre_municipio(df["municipio"])
        partes.append(pd.DataFrame({"codigo": cod, "nombre": nombre}).dropna())
    tabla = pd.concat(partes)
    return tabla.groupby("codigo")["nombre"].agg(
        lambda s: s.value_counts().idxmax()).to_dict()


def resolver_geografia(df, mapa_municipio):
    """Departamento y municipio resueltos desde el código (fuente confiable)."""
    codigo = limpiar_codigo_municipio(df["codigomunicipiohecho"])
    dep = aplicar_categoria(df["departamento"], CANONICOS_DEPARTAMENTO)
    dep = dep.where(codigo.isna(), codigo.map(
        lambda c: CODIGO_DEPARTAMENTO[c[:2]] if isinstance(c, str) else np.nan))
    mun = codigo.map(mapa_municipio).where(
        codigo.notna(), estandarizar_nombre_municipio(df["municipio"]))
    df = df.copy()
    df["departamento"] = dep
    df["municipio"] = mun
    df["codigomunicipiohecho"] = codigo
    return df
```

## 1.7 Paso 5 y 6: valores imposibles → faltantes, y faltantes → tratamiento

**Valores imposibles (se marcan como NaN, nunca se inventan):**
- Edad no entera, negativa o > 110 años.
- Año fuera de 2000–2026.
- ConteoRegistros ≤ 0 (columna crítica de cualquier conteo).

**Faltantes (se tratan al final, porque el paso 5 crea nuevos):**
- Categóricas → `"Sin información"` (marcar, no inventar).
- Edad → mediana. Año → derivado de la fecha o moda. ConteoRegistros → se descarta la fila.

```python
def marcar_imposibles(df):
    """Paso 5: marca como NaN todo valor imposible."""
    df = df.copy()
    if "edad" in df:
        edad = limpiar_numerico(df["edad"])
        df["edad"] = edad.where((edad >= 0) & (edad <= 110) & (edad == edad.round()))
    if "anohecho" in df:
        ano = limpiar_numerico(df["anohecho"])
        df["anohecho"] = ano.where((ano >= 2000) & (ano <= 2026))
    if "conteoregistros" in df:
        conteo = limpiar_numerico(df["conteoregistros"])
        df["conteoregistros"] = conteo.where(conteo > 0)
    return df


def imputar_faltantes(df, fuente):
    """Paso 6: imputa faltantes según el tipo de columna y la fuente."""
    df = df.copy()

    # Año: se deriva de una fecha confiable si existe; si no, por moda.
    if fuente == "Balance datos preliminares":
        modo_global = df["anohecho"].mode().iat[0] if not df["anohecho"].mode().empty else np.nan
        modo_version = df.groupby("fechaversion")["anohecho"].transform(
            lambda s: s.mode().iat[0] if not s.mode().empty else np.nan)
        df["anohecho"] = df["anohecho"].fillna(modo_version).fillna(modo_global)
        if df["fechaversion"].notna().any():
            df["fechaversion"] = df["fechaversion"].fillna(df["fechaversion"].mode().iat[0])
    elif fuente == "Festivos y recesos":
        df["anohecho"] = df["fechahecho"].dt.year.where(df["fechahecho"].notna())
        modo = df["anohecho"].mode().iat[0] if not df["anohecho"].mode().empty else np.nan
        df["anohecho"] = df["anohecho"].fillna(modo)
        # El día de la semana faltante se deriva de la fecha del hecho.
        df["diaocurrencia"] = df["diaocurrencia"].fillna(
            df["fechahecho"].dt.dayofweek.add(1).map(etiqueta_dia).where(df["fechahecho"].notna()))
    else:  # Comportamiento histórico
        modo = df["anohecho"].mode().iat[0] if not df["anohecho"].mode().empty else np.nan
        df["anohecho"] = df["anohecho"].fillna(modo)

    # Categóricas -> 'Sin información' (marcar, no inventar).
    categoricas = [c for c in ["mesocurrencia", "diaocurrencia", "rango3horas",
                               "zona", "sexo", "usuariovia", "actorvial",
                               "tipovehiculogrupo", "objetocolision",
                               "condicionvictima", "departamento", "municipio"]
                   if c in df.columns]
    for c in categoricas:
        if df[c].isna().any():
            df[c] = df[c].fillna("Sin información")

    # Edad -> mediana general.
    if "edad" in df and df["edad"].isna().any():
        df["edad"] = df["edad"].fillna(df["edad"].median())

    # ConteoRegistros es el corazón de cualquier conteo: si falta, se descarta la fila.
    df = df.dropna(subset=["conteoregistros"])
    return df
```

## 1.8 Pipeline de limpieza por archivo (los 9 pasos del cuaderno)

El procedimiento aplica, en orden, los pasos 1–7 del cuaderno a cada uno de los 3 archivos:
1. **Nombres de columnas** normalizados (minúsculas, espacios → `_`).
2. **Duplicados exactos** eliminados.
3. **Texto**: espacios al inicio/final y múltiples colapsados.
4. **Categorías** equivalentes (día, mes, rango, sexo, zona, usuario vía, geografía…).
5. **Imposibles → NaN**.
7. **Fechas** reales (`pd.to_datetime`).
6. **Faltantes** (después del 5, porque este crea nuevos).

```python
def limpiar_df(df, fuente, mapa_municipio):
    print(f"\n{'='*66}\nLIMPIEZA -> {fuente}\n{'='*66}")
    df = df.copy()

    # Paso 1: nombres de columnas (snake_case, minúsculas)
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
    print(f"Paso 1 - Columnas normalizadas ({df.shape[1]} columnas).")

    # Paso 2: duplicados exactos
    antes = len(df)
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Paso 2 - Duplicados eliminados: {antes - len(df)}")

    # Paso 3: texto - espacios al inicio/final y múltiples
    for c in df.select_dtypes(include=["object", "string"]).columns:
        df[c] = df[c].map(lambda v: " ".join(str(v).split()) if isinstance(v, str) else v)
    print("Paso 3 - Espacios colapsados en todas las columnas de texto.")

    # Paso 4: categorías equivalentes
    mapa_cats = {
        "diaocurrencia": lambda s: limpiar_dia(s),
        "mesocurrencia": lambda s: limpiar_mes(s),
        "rango3horas": lambda s: limpiar_rango(s),
        "zona": lambda s: aplicar_categoria(s, CANONICOS_ZONA),
        "sexo": lambda s: aplicar_categoria(s, CANONICOS_SEXO),
        "usuariovia": lambda s: aplicar_categoria(s, CANONICOS_USUARIO_VIA),
        "actorvial": lambda s: aplicar_categoria(s, CANONICOS_ACTOR_VIAL),
        "tipovehiculogrupo": lambda s: aplicar_categoria(s, CANONICOS_TIPO_VEHICULO, CASO_ESPECIAL_NA),
        "objetocolision": lambda s: aplicar_categoria(s, CANONICOS_OBJETO_COLISION),
        "condicionvictima": lambda s: aplicar_categoria(s, CANONICOS_CONDICION),
    }
    for col, fn in mapa_cats.items():
        if col in df.columns:
            df[col] = fn(df[col])
    df = resolver_geografia(df, mapa_municipio)
    n_cat = df["zona"].nunique() if "zona" in df else 0
    print(f"Paso 4 - Categorías estandarizadas (Zona->{n_cat} valores, Día->7, Mes->12, "
          f"geografía por código).")

    # Paso 5: valores imposibles -> faltantes
    n_antes = df["edad"].isna().sum() if "edad" in df else 0
    df = marcar_imposibles(df)
    if "edad" in df:
        print(f"Paso 5 - Imposibles marcados como faltantes "
              f"(Edad: +{df['edad'].isna().sum() - n_antes} nuevos).")

    # Paso 7: fechas reales
    if "fechaversion" in df:
        df["fechaversion"] = limpiar_fecha(df["fechaversion"])
    if "fechahecho" in df:
        df["fechahecho"] = limpiar_fecha(df["fechahecho"])

    # Paso 6: faltantes (después del 5, que crea nuevos)
    df = imputar_faltantes(df, fuente)

    # Columna de trazabilidad
    df["fuente"] = fuente
    print(f"-> {fuente}: {df.shape[0]} filas limpias.")
    return df
```

# PUNTO 2 — Selección de 2 modelos y justificación

El problema es **clasificación binaria supervisada** (Urbana/Rural), con ~99 mil registros,
mayoría de atributos categóricos nominales (10 de 13 columnas) y clases proporcionalmente
balanceadas (59%/41%). Se eligen **dos modelos de naturaleza distinta** para comparar un
enfoque lineal e interpretable contra uno no lineal de conjunto:

## 2.1 Modelo 1: Regresión Logística

### Qué problema resuelve
Clasificación binaria mediante una frontera lineal en el espacio de atributos. Estima la
**probabilidad** de que el siniestro sea urbano:

$$P(\text{Urbana}) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + \dots + \beta_k x_k)}}$$

### Justificación de elección
| Criterio | Cómo aplica a este caso |
|---|---|
| **Problema binario** | La zona tiene exactamente 2 clases (Urbana/Rural): es el caso ideal de la logística. |
| **Interpretabilidad** | Da coeficientes con signo y magnitud: se puede responder *"¿qué lleva a un siniestro a ser más urbano?"*. Muy valioso en una entrega académica. |
| **Eficiencia y estabilidad** | Con ~99 mil filas y ~112 atributos codificados (imputación + OneHot), es rápido de entrenar y la validación cruzada no se vuelve costosa. |
| **Baseline estándar** | El cuaderno del curso la usa como modelo de referencia de clasificación binaria (sección 4 del Panorama). |
| **Limitación asumida** | No captura no linealidades ni interacciones entre departamento × condición de la víctima; por eso se compara con un modelo no lineal. |

## 2.2 Modelo 2: Random Forest

### Qué problema resuelve
Modelo de **conjunto**: entrena cientos de árboles de decisión sobre muestras aleatorias de
filas (bootstrap) y columnas (`max_features`), y combina sus votos. Captura fronteras no
lineales e **interacciones** automáticamente, sin exigir escalamiento.

### Justificación de su elección
| Criterio | Cómo aplica a este caso |
|---|---|
| **Atributos categóricos nominales** | Los árboles dividen por categoría y descubren reglas como *"departamento=Atlántico → salta"*, sin que la linealidad lo limite. |
| **No exige escalamiento** | Edad y año conviven sin escalar; OneHot + `handle_unknown="ignore"` es suficiente. |
| **Robustez y generalización** | Es robusto a ruido y atípicos (los 1.5×IQR de la edad), y su ensamblado reduce la varianza frente a un árbol individual. Es el modelo de conjunto del Panorama (secciones 8 y 9). |
| **Importancia de variables** | Proporciona `feature_importances_`, útil para concluir qué explica la zona. |
| **Limitación asumida** | Menos interpretable que la logística; se mitiga con la importancia de variables agrupada por variable original. |

## 2.3 Comparación en una tabla

| Criterio | Regresión Logística | Random Forest |
|---|---|---|
| Tipo de frontera | Lineal | No lineal (segmentada) |
| Interpretabilidad | Alta (coeficientes/odds) | Media (importancias) |
| Requiere escalamiento | Sí (StandardScaler) | No |
| Interacciones entre variables | No | Sí, automáticas |
| Riesgo principal | Bajo (regularizado) | Sobreajuste si demasiados árboles/profundidad |
| Métricas de interés | AUC, precisión/recall | AUC, F1, importancias |

> **Conclusión de la selección:** dos enfoques complementarios. Si gana la logística, la
> frontera es esencialmente lineal; si gana el Random Forest, existen interacciones que la
> logística no ve. La métrica decisiva será el **ROC-AUC** en test (no el accuracy, porque
> hay 18 puntos de desbalance).

# Preparación de datos para Machine Learning

- **Variable objetivo (y):** `zona` → 1 si `Urbana`, 0 si `Rural`.
- Las filas con `zona == "Sin información"` **se eliminan** para entrenar: no tienen etiqueta.
  El resto de "Sin información" en los *atributos* se conserva como categoría (el modelo
  aprende a tratarla).
- `train_test_split` con `stratify` para conservar la misma proporción Urbana/Rural en
  entrenamiento y prueba.

```python
# Dataset de modelado: solo filas con etiqueta de zona válida
df_modelo = df_limpio[df_limpio["zona"].isin(["Urbana", "Rural"])].copy()
print(f"Filas con zona válida (etiqueta): {len(df_modelo)}")
print(df_modelo["zona"].value_counts(), "\n")

# Columnas que no son información de predicción utilizable: se descartan
df_modelo = df_modelo.drop(columns=["fechaversion", "fechahecho",
                                    "municipio", "codigomunicipiohecho"])

columnas_cat = ["mesocurrencia", "diaocurrencia", "rango3horas", "departamento",
                "sexo", "usuariovia", "actorvial", "tipovehiculogrupo",
                "objetocolision", "condicionvictima"]
columnas_num = ["anohecho", "edad", "es_festivo_receso"]

X = df_modelo[columnas_cat + columnas_num]
y = (df_modelo["zona"] == "Urbana").astype(int)  # 1 = Urbana, 0 = Rural

print("Distribución del objetivo (1 = Urbana):")
print(y.value_counts().to_string())
print(f"\nProporción Urbana: {y.mean():.3f}  ·  Rural: {1 - y.mean():.3f}")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=SEED)
print(f"\nEntrenamiento: {X_train.shape[0]} filas  ·  Test: {X_test.shape[0]} filas")
```

# PUNTO 3 — Aplicación de los modelos

## 3.1 Preprocesador común en un `ColumnTransformer`

Todo vive dentro de un `Pipeline` → lo que se aprende (imputación, escalado, OneHot) se
ajusta **solo sobre el conjunto de entrenamiento** y se reutiliza en el test. Esto evita la
fuga de información y es la misma filosofía de las secciones 4 y 9 del Panorama.

- **Numéricas** (`anohecho`, `edad`, `es_festivo_receso`): imputación con mediana +
  `StandardScaler` (necesario para la logística).
- **Categóricas** (10 columnas): imputación con constante `"Sin información"` +
  `OneHotEncoder` con `handle_unknown="ignore"` (categorías no vistas en entrenamiento no
  rompen el modelo).

```python
preprocesador = ColumnTransformer(
    transformers=[
        ("num", Pipeline([
            ("imputar", SimpleImputer(strategy="median")),
            ("escalar", StandardScaler()),
        ]), columnas_num),
        ("cat", Pipeline([
            ("imputar", SimpleImputer(strategy="constant", fill_value="Sin información")),
            ("codificar", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), columnas_cat),
    ]
)

modelo_logit = Pipeline([
    ("prep", preprocesador),
    ("clf", LogisticRegression(max_iter=2000, C=1.0, random_state=SEED)),
])

modelo_rf = Pipeline([
    ("prep", preprocesador),
    ("clf", RandomForestClassifier(n_estimators=200, random_state=SEED, n_jobs=-1)),
])
print("Pipelines listos.")
print("Atributos categóricos:", len(columnas_cat), "| numéricos:", len(columnas_num))
```

## 3.2 Validación cruzada (Stratified 5-Fold)

Un único split depende de qué filas cayeron al azar en cada partición. Con **5 pliegues
estratificados** cada modelo se entrena 5 veces y se reporta la media ± desviación, tanto en
ROC-AUC como en accuracy. Así la comparación es robusta.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
resultados_cv = {}

for nombre, modelo in [("Regresión Logística", modelo_logit), ("Random Forest", modelo_rf)]:
    auc = cross_val_score(modelo, X_train, y_train, cv=cv, scoring="roc_auc", n_jobs=-1)
    acc = cross_val_score(modelo, X_train, y_train, cv=cv, scoring="accuracy", n_jobs=-1)
    resultados_cv[nombre] = {"roc_auc": auc, "accuracy": acc}
    print(f"{nombre}")
    print(f"  ROC-AUC : {auc.mean():.4f} ± {auc.std():.4f}")
    print(f"  Accuracy: {acc.mean():.4f} ± {acc.std():.4f}")
```

### Lectura de la validación cruzada

- Ambos modelos superan claramente al azar (AUC ≈ 0.5). El ROC-AUC queda en **~0.73–0.75**:
  la zona se explica sobre todo por geografía (departamento) y la señal horaria/vehicular es
  más débil.
- La desviación entre pliegues es pequeña (desv. estándar ≈ 0.002–0.005) → resultados estables.
- **La Regresión Logística supera ligeramente al Random Forest** en AUC y en accuracy. Eso
  indica que la frontera Urbana/Rural es **esencialmente lineal**: la capacidad extra del
  Random Forest (interacciones no lineales) no se traduce en una mejor generalización, e
  incluso añade algo de varianza. Es el caso clásico en el que el modelo más simple gana.

```python
# Entrenamiento final sobre todo el set de entrenamiento y evaluación en test
mejores = {}
for nombre, modelo in [("Regresión Logística", modelo_logit), ("Random Forest", modelo_rf)]:
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)
    y_prob = modelo.predict_proba(X_test)[:, 1]
    mejores[nombre] = (y_test, y_pred, y_prob)
    print(f"\n### {nombre}")
    print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_test, y_pred):.4f}")
    print(f"F1       : {f1_score(y_test, y_pred):.4f}")
    print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob):.4f}")
    print(classification_report(y_test, y_pred, target_names=["Rural (0)", "Urbana (1)"]))
```

```python
# -*- coding: utf-8 -*-
"""EDA + análisis estadístico + hallazgos sobre siniestros_viales_limpio.csv"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, kruskal
from IPython.display import display

import os

sns.set_style("whitegrid")
FIGDIR = "/content/figs"
os.makedirs(FIGDIR, exist_ok=True)

df = pd.read_csv("siniestros_viales_limpio.csv", sep=";", low_memory=False)
print("Filas:", df.shape[0], "Columnas:", df.shape[1])

TOTAL_FALLECIDOS = df["conteoregistros"].sum()
print(f"Total de personas fallecidas (ponderado): {int(TOTAL_FALLECIDOS):,}")

# ------------------------------------------------------------------
# Helpers
# ------------------------------------------------------------------
def distribucion_ponderada(df, columna, top=15):
    d = df.dropna(subset=[columna])
    tabla = d.groupby(columna)["conteoregistros"].sum().sort_values(ascending=False)
    pct = (tabla / tabla.sum() * 100).round(2)
    return pd.DataFrame({"fallecidos": tabla, "%": pct}).head(top)

def cramers_v(df, col1, col2, pesos="conteoregistros"):
    d = df.dropna(subset=[col1, col2])
    tabla = pd.pivot_table(d, values=pesos, index=col1, columns=col2, aggfunc="sum", fill_value=0)
    chi2, p, gl, _ = chi2_contingency(tabla)
    n = tabla.values.sum()
    k = min(tabla.shape) - 1
    v = np.sqrt(chi2 / (n * k)) if k > 0 else np.nan
    return chi2, p, gl, v

def prueba_chi2(df, col1, col2, alfa=0.05):
    chi2, p, gl, v = cramers_v(df, col1, col2)
    concl = "SÍ hay asociación (se rechaza H0)" if p < alfa else "sin evidencia de asociación"
    print(f"  {col1} vs {col2}: chi2={chi2:.1f}  gl={gl}  p={p:.4g}  V Cramér={v:.3f}  -> {concl}")
    return chi2, p, v

# ==================================================================
# 1. EDA - Distribuciones univariantes
# ==================================================================
print("\n" + "="*60 + "\n1. EDA - DISTRIBUCIONES UNIVARIANTES\n" + "="*60)

zona_dist = distribucion_ponderada(df, "zona")
sexo_dist = distribucion_ponderada(df, "sexo")
usuario_dist = distribucion_ponderada(df, "usuariovia")
vehiculo_dist = distribucion_ponderada(df, "tipovehiculogrupo")
depto_dist = distribucion_ponderada(df, "departamento", top=10)
anio_dist = df.groupby("anohecho")["conteoregistros"].sum().sort_index()
dia_dist = df.dropna(subset=["diaocurrencia"]).groupby("diaocurrencia")["conteoregistros"].sum()
franja_dist = df.dropna(subset=["rango3horas"]).groupby("rango3horas")["conteoregistros"].sum()

print("\nZona:\n", zona_dist)
print("\nSexo:\n", sexo_dist)
print("\nUsuario de la vía:\n", usuario_dist)
print("\nTipo de vehículo:\n", vehiculo_dist)
print("\nTop 10 departamentos:\n", depto_dist)
print("\nPor año:\n", anio_dist)

df_edad = df.dropna(subset=["edad"]).copy()
edades_ponderadas = np.repeat(df_edad["edad"].values, df_edad["conteoregistros"].astype(int).values)
print("\nEstadísticos de edad (ponderados):\n", pd.Series(edades_ponderadas).describe())

# --- Figura 1: panel univariante ---
fig, axes = plt.subplots(2, 3, figsize=(19, 10))

zona_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[0,0], color="#003366")
axes[0,0].set_title("Fallecidos por zona"); axes[0,0].set_xlabel("Fallecidos")

sexo_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[0,1], color="#0059b3")
axes[0,1].set_title("Fallecidos por sexo"); axes[0,1].set_xlabel("Fallecidos")

sns.histplot(edades_ponderadas, bins=30, kde=True, color="#003366", ax=axes[0,2])
axes[0,2].axvline(np.median(edades_ponderadas), color="red", linestyle="--", label="Mediana")
axes[0,2].set_title("Distribución de edad"); axes[0,2].legend()

usuario_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,0], color="#336699")
axes[1,0].set_title("Fallecidos por usuario de la vía"); axes[1,0].set_xlabel("Fallecidos")

vehiculo_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,1], color="#6699cc")
axes[1,1].set_title("Fallecidos por tipo de vehículo"); axes[1,1].set_xlabel("Fallecidos")

depto_dist["fallecidos"].sort_values().plot(kind="barh", ax=axes[1,2], color="#00264d")
axes[1,2].set_title("Top 10 departamentos"); axes[1,2].set_xlabel("Fallecidos")

plt.tight_layout()
plt.savefig(f"{FIGDIR}/01_distribuciones_univariantes.png", dpi=130)
display(fig)
plt.close(fig)

# --- Figura 2: evolución temporal ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
anio_dist.plot(kind="line", marker="o", ax=axes[0], color="#003366")
axes[0].set_title("Evolución anual de fallecidos"); axes[0].set_xlabel("Año"); axes[0].set_ylabel("Fallecidos")

dia_dist.reindex(sorted(dia_dist.index)).plot(kind="bar", ax=axes[1], color="#336699")
axes[1].set_title("Fallecidos por día de la semana"); axes[1].tick_params(axis='x', rotation=45)

franja_dist.plot(kind="bar", ax=axes[2], color="#6699cc")
axes[2].set_title("Fallecidos por franja horaria"); axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig(f"{FIGDIR}/02_evolucion_temporal.png", dpi=130)
display(fig)
plt.close(fig)

# ==================================================================
# 2. EDA - Relaciones bivariantes
# ==================================================================
print("\n" + "="*60 + "\n2. EDA - RELACIONES BIVARIANTES\n" + "="*60)

def tabla_cruzada_pct(df, fila, columna):
    d = df.dropna(subset=[fila, columna])
    tabla = pd.pivot_table(d, values="conteoregistros", index=fila, columns=columna, aggfunc="sum", fill_value=0)
    return tabla.div(tabla.sum(axis=1), axis=0).mul(100).round(1)

zona_usuario = tabla_cruzada_pct(df, "zona", "usuariovia")
sexo_usuario = tabla_cruzada_pct(df, "sexo", "usuariovia")
print("\nZona x Usuario de la vía (% fila):\n", zona_usuario)
print("\nSexo x Usuario de la vía (% fila):\n", sexo_usuario)

df_pond = df.dropna(subset=["edad", "zona", "sexo"]).copy()
df_pond = df_pond.loc[df_pond.index.repeat(df_pond["conteoregistros"].astype(int))]

dia_franja = None
if {"diaocurrencia", "rango3horas"}.issubset(df.columns):
    d = df.dropna(subset=["diaocurrencia", "rango3horas"])
    dia_franja = pd.pivot_table(d, values="conteoregistros", index="diaocurrencia",
                                 columns="rango3horas", aggfunc="sum", fill_value=0)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

sns.heatmap(zona_usuario, annot=True, fmt=".1f", cmap="Blues", ax=axes[0,0])
axes[0,0].set_title("% usuario de la vía dentro de cada zona")

sns.boxplot(data=df_pond, x="zona", y="edad", palette="Blues", ax=axes[0,1])
axes[0,1].set_title("Edad por zona")

sns.boxplot(data=df_pond, x="sexo", y="edad", palette="Blues", ax=axes[1,0])
axes[1,0].set_title("Edad por sexo")

if dia_franja is not None:
    dia_franja_plot = dia_franja.drop(index="Sin información", columns="Sin información", errors="ignore")
    sns.heatmap(dia_franja_plot.reindex(sorted(dia_franja_plot.index)), cmap="Reds", ax=axes[1,1])
    axes[1,1].set_title("Fallecidos: día de la semana x franja horaria\n(excluye registros 'Sin información')")

plt.tight_layout()
plt.savefig(f"{FIGDIR}/03_relaciones_bivariantes.png", dpi=130)
display(fig)
plt.close(fig)

# ==================================================================
# 3. Fuerza de asociación (V de Cramér) entre categóricas
# ==================================================================
print("\n" + "="*60 + "\n3. V DE CRAMÉR ENTRE VARIABLES CATEGÓRICAS\n" + "="*60)

VARS_CAT = [c for c in ["zona", "sexo", "usuariovia", "tipovehiculogrupo",
                         "objetocolision", "condicionvictima", "departamento",
                         "diaocurrencia", "rango3horas"] if c in df.columns]

matriz_v = pd.DataFrame(index=VARS_CAT, columns=VARS_CAT, dtype=float)
for a in VARS_CAT:
    for b in VARS_CAT:
        if a == b:
            matriz_v.loc[a, b] = 1.0
        else:
            try:
                _, _, _, v = cramers_v(df, a, b)
                matriz_v.loc[a, b] = v
            except Exception:
                matriz_v.loc[a, b] = np.nan

print(matriz_v.round(3))

fig = plt.figure(figsize=(9, 7))
sns.heatmap(matriz_v.astype(float), annot=True, fmt=".2f", cmap="viridis", vmin=0, vmax=1)
plt.title("V de Cramér entre variables categóricas")
plt.tight_layout()
plt.savefig(f"{FIGDIR}/04_cramers_v.png", dpi=130)
display(fig)
plt.close(fig)

# ==================================================================
# 4. Análisis estadístico - Pruebas de hipótesis (alfa = 0.05)
# ==================================================================
print("\n" + "="*60 + "\n4. PRUEBAS DE HIPÓTESIS (alfa = 0.05)\n" + "="*60)

print("\nChi-cuadrado de independencia:")
r_zona_usuario = prueba_chi2(df, "zona", "usuariovia")
r_zona_vehiculo = prueba_chi2(df, "zona", "tipovehiculogrupo")
r_sexo_usuario = prueba_chi2(df, "sexo", "usuariovia")
r_zona_sexo = prueba_chi2(df, "zona", "sexo")
if "diaocurrencia" in df and "rango3horas" in df:
    r_dia_franja = prueba_chi2(df, "diaocurrencia", "rango3horas")

print("\nKruskal-Wallis (¿difiere la edad entre grupos?):")
grupos_zona = [df_pond.loc[df_pond["zona"] == z, "edad"] for z in df_pond["zona"].unique()]
h_zona, p_zona = kruskal(*grupos_zona)
print(f"  Edad según zona: H={h_zona:.1f}  p={p_zona:.4g}  -> "
      f"{'diferencia significativa' if p_zona < 0.05 else 'sin diferencia significativa'}")

grupos_sexo = [df_pond.loc[df_pond["sexo"] == s, "edad"] for s in df_pond["sexo"].unique()]
h_sexo, p_sexo = kruskal(*grupos_sexo)
print(f"  Edad según sexo: H={h_sexo:.1f}  p={p_sexo:.4g}  -> "
      f"{'diferencia significativa' if p_sexo < 0.05 else 'sin diferencia significativa'}")

df_pond_u = df.dropna(subset=["edad", "usuariovia"]).copy()
df_pond_u = df_pond_u.loc[df_pond_u.index.repeat(df_pond_u["conteoregistros"].astype(int))]
grupos_usuario = [df_pond_u.loc[df_pond_u["usuariovia"] == u, "edad"] for u in df_pond_u["usuariovia"].unique()]
h_usr, p_usr = kruskal(*grupos_usuario)
print(f"  Edad según usuario de la vía: H={h_usr:.1f}  p={p_usr:.4g}  -> "
      f"{'diferencia significativa' if p_usr < 0.05 else 'sin diferencia significativa'}")

# ==================================================================
# 5. Hallazgos, patrones e insights (generados desde los datos reales)
# ==================================================================
print("\n" + "="*60 + "\n5. HALLAZGOS\n" + "="*60)

top_departamento = depto_dist.index[0]; pct_top_departamento = depto_dist["%"].iloc[0]
zona_mayoritaria = zona_dist.index[0]; pct_zona_mayoritaria = zona_dist["%"].iloc[0]
usuario_mayoritario = usuario_dist.index[0]; pct_usuario_mayoritario = usuario_dist["%"].iloc[0]
sexo_mayoritario = sexo_dist.index[0]; pct_sexo_mayoritario = sexo_dist["%"].iloc[0]
vehiculo_mayoritario = vehiculo_dist.index[0]; pct_vehiculo_mayoritario = vehiculo_dist["%"].iloc[0]
edad_mediana = np.median(edades_ponderadas)
anio_pico = anio_dist.idxmax(); fallecidos_anio_pico = int(anio_dist.max())
dia_pico = dia_dist.drop(labels=["Sin información"], errors="ignore").idxmax() if not dia_dist.empty else None
franja_pico = franja_dist.drop(labels=["Sin información"], errors="ignore").idxmax() if not franja_dist.empty else None
zona_usuario_top_urbana = zona_usuario.loc["Urbana"].idxmax() if "Urbana" in zona_usuario.index else None
zona_usuario_top_rural = zona_usuario.loc["Rural"].idxmax() if "Rural" in zona_usuario.index else None

hallazgos = f"""
1. CONCENTRACIÓN GEOGRÁFICA: {top_departamento} concentra el {pct_top_departamento}% de los
   fallecidos del dataset ({depto_dist['fallecidos'].iloc[0]:,.0f} de {TOTAL_FALLECIDOS:,.0f}).

2. ZONA: el {pct_zona_mayoritaria}% de los fallecidos ocurre en zona {str(zona_mayoritaria).lower()}.
   Chi-cuadrado zona-usuariovia: p={r_zona_usuario[1]:.4g}, V Cramér={r_zona_usuario[2]:.3f}
   ({'asociación' if r_zona_usuario[1] < 0.05 else 'sin asociación'} estadísticamente significativa).
   En zona urbana el perfil dominante es "{zona_usuario_top_urbana}"; en zona rural, "{zona_usuario_top_rural}".

3. ACTOR VIAL MÁS VULNERABLE: "{usuario_mayoritario}" representa el {pct_usuario_mayoritario}%
   de los fallecidos, el grupo de mayor riesgo. El vehículo más involucrado es
   "{vehiculo_mayoritario}" ({pct_vehiculo_mayoritario}%).

4. SEXO: el {pct_sexo_mayoritario}% de los fallecidos es "{sexo_mayoritario}".
   Kruskal-Wallis edad~sexo: H={h_sexo:.1f}, p={p_sexo:.4g}
   ({'diferencia' if p_sexo < 0.05 else 'sin diferencia'} significativa de edad entre sexos).

5. EDAD: mediana de {edad_mediana:.0f} años. Kruskal-Wallis edad~zona: p={p_zona:.4g};
   edad~usuario de la vía: p={p_usr:.4g}.

6. TENDENCIA TEMPORAL: {int(anio_pico)} es el año con más fallecidos ({fallecidos_anio_pico:,}).
   Día con más fallecidos: {dia_pico}. Franja horaria más crítica (excluyendo "Sin información",
   que representa el {franja_dist.get('Sin información', 0) / franja_dist.sum() * 100:.1f}% de
   los registros con franja): {franja_pico}.
"""
print(hallazgos)

with open(f"{FIGDIR}/hallazgos.txt", "w", encoding="utf-8") as f:
    f.write(hallazgos)

print("\nListo. Figuras guardadas en", FIGDIR)
```

## Conclusiones y Recomendaciones

### Conclusiones Clave

1.  **Concentración Geográfica y Zonal:** Los siniestros viales fatales se concentran significativamente en departamentos como Antioquia, Valle del Cauca, Cundinamarca y Bogotá D.C. La mayoría de estos incidentes ocurren en zonas urbanas (aproximadamente el 56.71%), donde los usuarios de moto y peatones son los más afectados. Existe una asociación estadísticamente significativa entre la zona (urbana/rural) y el tipo de usuario de la vía involucrado.

2.  **Actor Vial Más Vulnerable:** Los **usuarios de moto** son el grupo más vulnerable, representando más de la mitad (56.67%) de los fallecidos, y las motocicletas son el tipo de vehículo más presente en los siniestros fatales. Los peatones también constituyen una parte significativa de las víctimas (22.81%).

3.  **Perfil Demográfico:** La mayoría de las víctimas son hombres (80.68%). La edad mediana de los fallecidos es de 41 años, y se observan diferencias significativas en la distribución de la edad según la zona y el tipo de usuario de la vía.

4.  **Tendencia Temporal:** El año 2024 ha registrado el mayor número de fallecidos en el período analizado. Aunque no se detallan días y franjas horarias específicas en el resumen, el análisis revela picos de accidentalidad en ciertos días de la semana y rangos horarios.

5.  **Rendimiento del Modelo Predictivo:** La Regresión Logística demostró ser el modelo más efectivo para clasificar la zona de ocurrencia de los siniestros (urbana/rural), con un ROC-AUC de 0.7524. Esto sugiere que la relación entre las características de los siniestros y su ubicación (urbana/rural) es predominantemente lineal. Sin embargo, el desempeño no es perfecto, lo que indica que existen otros factores no capturados por los datos actuales que influyen en la ocurrencia de siniestros.

### Recomendaciones

1.  **Intervenciones Focalizadas Geográficamente:** Implementar campañas de seguridad vial e infraestructura en los departamentos con mayor siniestralidad, como Antioquia, Valle del Cauca, Cundinamarca y Bogotá D.C., con un enfoque específico en las zonas urbanas.

2.  **Priorizar la Seguridad del Motociclista y Peatón:** Desarrollar programas de educación, fiscalización y mejoramiento de la infraestructura vial dirigidos específicamente a motociclistas y peatones, los cuales son los actores viales más vulnerables.

3.  **Campañas Demográficas Específicas:** Diseñar campañas de concientización y prevención que consideren el perfil demográfico de las víctimas, como campañas dirigidas a hombres y a grupos de edad específicos, especialmente aquellos alrededor de la mediana de 41 años.

4.  **Análisis y Acciones Temporales:** Investigar a fondo los días y franjas horarias con mayor incidencia de siniestros para establecer medidas preventivas, como el aumento de la presencia policial, campañas de sensibilización en momentos críticos o ajustes en la señalización.

5.  **Exploración de Variables Adicionales:** Dado que los modelos no explican completamente la ocurrencia de siniestros, se recomienda investigar y recolectar datos sobre factores adicionales que podrían influir, como condiciones climáticas, estado de las vías, comportamiento de los conductores (distracciones, velocidad) o tipos específicos de infraestructura vial (curvas peligrosas, intersecciones).

6.  **Monitoreo Continuo:** Mantener un monitoreo constante de los patrones de siniestralidad vial para evaluar la efectividad de las intervenciones y ajustar las estrategias según sea necesario. La simplicidad y interpretabilidad de la Regresión Logística facilitarán este monitoreo.
