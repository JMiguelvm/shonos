## A1. Tipos de datos primitivos

Todo valor en Python tiene un **tipo**. La función `type()` nos dice cuál es — es la primera pregunta que un científico de datos le hace a un dato desconocido.

```python
# Los cuatro tipos básicos que usarás todos los días
edad = 21                    # int    -> números enteros
estatura = 1.68              # float  -> números con decimales
nombre = "Valentina"         # str    -> texto (cadenas de caracteres)
es_estudiante = True         # bool   -> verdadero o falso

print(type(edad))
print(type(estatura))
print(type(nombre))
print(type(es_estudiante))
```

### `int` (enteros)

Los enteros son números completos, positivos o negativos, **sin decimales**. En Python tienen *precisión arbitraria*: pueden ser tan grandes como la memoria lo permita.

**Usos típicos:**
- **Conteo:** número de estudiantes, productos en inventario, veces que ocurre un evento.
- **Identificadores:** IDs de usuarios, números de registro.
- **Índices:** para acceder a elementos en listas y cadenas.
- **Operaciones matemáticas:** sumas, restas, multiplicaciones, divisiones enteras.

```python
numero_estudiantes = 35
temperatura_minima = -5      # también pueden ser negativos

# Precisión arbitraria: Python aguanta números gigantes sin despeinarse
numero_gigante = 2 ** 100
print(numero_gigante)
print(type(numero_gigante))   # sigue siendo int
```

### `float` (números de punto flotante)

Los flotantes son números reales que tienen **parte decimal**. Se utilizan para representar valores que no son enteros.

**Usos típicos:**
- **Mediciones:** temperaturas, alturas, pesos, distancias.
- **Cálculos científicos y de ingeniería.**
- **Valores monetarios y precios.**
- **Promedios y porcentajes.**

```python
estatura_metros = 1.72
precio_cafe = 2500.0     # el .0 lo convierte en float aunque sea "redondo"
print(type(estatura_metros), type(precio_cafe))

# ⚠️ Los floats no son exactos: la computadora los guarda en binario
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)   # ¡False! Esto sorprende a todo el mundo

# Por eso en análisis se compara con una tolerancia:
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
```

### `str` (cadenas de texto)

Las cadenas son **secuencias inmutables de caracteres**. Se usan para almacenar texto: nombres, mensajes, direcciones, categorías.

**Usos típicos:**
- **Nombres y etiquetas.**
- **Mensajes y descripciones.**
- **Datos categóricos** en análisis (ciudad, programa, género musical...).
- **Rutas de archivos y URLs.**

```python
universidad = "Universidad EAN"

print(len(universidad))          # cuántos caracteres tiene
print(universidad.upper())       # TODO EN MAYÚSCULAS
print(universidad.lower())       # todo en minúsculas
print(universidad[0])            # el primer carácter (¡también se indexa desde 0!)

# strip() quita espacios sobrantes: clave para limpiar datos
ciudad_sucia = "  Bogotá  "
print(ciudad_sucia.strip())
```

### `bool` (booleanos)

Los booleanos representan uno de dos valores: `True` (verdadero) o `False` (falso). Son la base de todas las **decisiones** en programación.

**Usos típicos:**
- **Condiciones lógicas** (¿aprobó?, ¿es mayor de edad?, ¿pagó con la app?).
- **Banderas de estado** (activo/inactivo).
- **Resultados de comparaciones** — toda comparación devuelve un booleano.

### `None` (ausencia de valor)

`None` representa la **ausencia** de un valor: no es cero, no es texto vacío, es *nada*. En los datos reales aparecerá muchísimo como dato faltante.

```python
mayor_de_edad = True
print(type(mayor_de_edad))

# Toda comparación produce un booleano:
print(5 > 3)
print(10 == 20)

# None: la ausencia de valor
celular = None
print(type(celular))
print(celular is None)
```

## A2. Conversión de tipos (*casting*)

Podemos convertir un valor de un tipo a otro con `int()`, `float()`, `str()` y `bool()`. Esto será **el corazón de la limpieza de datos** en las próximas semanas: los datos casi siempre llegan como texto y hay que convertirlos.

```python
# Conversiones que funcionan
print(int("25"))        # texto -> entero
print(float("3.14"))    # texto -> decimal
print(str(2026))        # número -> texto
print(int(9.99))        # float -> int: ¡TRUNCA, no redondea!
print(round(9.99))      # para redondear se usa round()

# bool() tiene su propia lógica: 0 es False, cualquier otro número es True
print(bool(0))
print(bool(100))
print(bool(""))         # texto vacío -> False
print(bool("hola"))     # texto con contenido -> True
```

```python
# Conversiones que FALLAN (ejecuta y lee el error con calma)
int("25 años")
```

```python
# En Colombia escribimos 1.234,56 — Python espera 1234.56
# Este es un problema REAL que verás en datasets colombianos:
precio_texto = "1.234,56"

# float(precio_texto)  # <- esto fallaría. Hay que limpiar primero:
precio_limpio = precio_texto.replace(".", "").replace(",", ".")
precio = float(precio_limpio)
print(precio, type(precio))
```

## A3. Operadores y jerarquía de operaciones

Los **operadores** son símbolos especiales que realizan operaciones sobre uno o más valores (llamados *operandos*). Comprender cómo se procesan los cálculos es esencial para escribir programas correctos. Python los clasifica en varias categorías:

### Categorías de operadores

| Categoría | Operador | Descripción | Ejemplo y resultado |
| :-------- | :------- | :---------- | :------------------ |
| **Aritméticos** | `+` | Suma | `7 + 3  # 10` |
| | `-` | Resta | `7 - 3  # 4` |
| | `*` | Multiplicación | `7 * 3  # 21` |
| | `/` | División (flotante) | `9 / 2  # 4.5` |
| | `//` | División entera | `9 // 2  # 4` |
| | `%` | Módulo (resto) | `9 % 2  # 1` |
| | `**` | Exponenciación | `2 ** 3  # 8` |
| **Comparación** | `==` | Igual a | `5 == 3  # False` |
| | `!=` | Diferente de | `5 != 3  # True` |
| | `<` | Menor que | `5 < 3  # False` |
| | `>` | Mayor que | `5 > 3  # True` |
| | `<=` | Menor o igual que | `5 <= 5  # True` |
| | `>=` | Mayor o igual que | `5 >= 3  # True` |
| **Lógicos** | `and` | Y lógico (ambas verdaderas) | `True and False  # False` |
| | `or` | O lógico (una o ambas) | `True or False  # True` |
| | `not` | Negación lógica | `not True  # False` |

> 📌 Los operadores de **comparación** siempre devuelven un booleano (`True` o `False`).
> 📌 Los operadores **lógicos** se utilizan para combinar expresiones booleanas.

```python
# ARITMÉTICOS: las tres divisiones responden preguntas distintas
print(9 / 2)     # 4.5 -> ¿cuánto da exactamente?      (siempre float)
print(9 // 2)    # 4   -> ¿cuántas veces cabe completo?
print(9 % 2)     # 1   -> ¿cuánto sobra?               (par/impar, ciclos)
print(2 ** 10)   # 1024 -> potencias

# ⚠️ Y recuerda: el operador obedece al TIPO del dato
print(5 + 3)         # suma de enteros -> 8
print("5" + "3")     # concatenación de textos -> "53" 😱
print("ja" * 3)      # repetición de texto -> "jajaja"
```

```python
# COMPARACIÓN: siempre devuelven True o False
nota_final = 3.8
nota_aprobatoria = 3.0

print(nota_final >= nota_aprobatoria)   # ¿aprobó?
print(nota_final == 5.0)                # ¿fue nota perfecta?
print(nota_final != 0.0)                # ¿es distinta de cero?

# ⚠️ No confundas:  =  asigna un valor,  ==  compara dos valores
```

```python
# LÓGICOS: combinan condiciones, como en la vida real
promedio = 4.2
semestre = 4
estrato = 2

# ¿Cumple los requisitos de la beca? (las DOS condiciones deben cumplirse)
print(promedio >= 4.0 and semestre >= 3)

# ¿Aplica al descuento? (basta con UNA de las dos)
print(semestre == 1 or estrato <= 2)

# not invierte el resultado
print(not promedio >= 4.0)
```

### Jerarquía de operaciones: la Regla de Oro

La **jerarquía de operaciones** (orden de operaciones o *precedencia*) define el orden en que se evalúan las operaciones dentro de una expresión. Seguirla es crucial para obtener el resultado esperado.

El orden de precedencia en Python, **de mayor a menor**:

| Prioridad | Operadores | Detalle |
|:---:|---|---|
| 1 | `( )` | Lo que está entre paréntesis se evalúa **primero** |
| 2 | `**` | Exponenciación (potencias) |
| 3 | `*`, `/`, `//`, `%` | Multiplicación y divisiones, de izquierda a derecha |
| 4 | `+`, `-` | Suma y resta, de izquierda a derecha |
| 5 | `==`, `!=`, `<`, `>`, `<=`, `>=` | Comparaciones (después de la aritmética) |
| 6 | `not` → `and` → `or` | Lógicos: `not` primero, `or` de último |

> 🥇 **La Regla de Oro:** ante la duda, **usa paréntesis**. No cuestan nada, eliminan la ambigüedad y hacen que cualquier persona (incluido tu yo del futuro) lea la fórmula sin adivinar.

```python
# ¿Coinciden con lo que esperabas? Piensa ANTES de ejecutar
print(10 + 2 * 3 - 1)        # ¿26? ¿15? ...
print(10 / 2 + 3 ** 2)       # las potencias van antes que la división
print((10 + 2) * (3 - 1))    # los paréntesis mandan
print(True and (5 > 3) or False)
```

```python
# 🚨 LA TRAMPA CLÁSICA: el promedio mal calculado
nota_corte1 = 4.0
nota_corte2 = 3.0

promedio_mal = nota_corte1 + nota_corte2 / 2      # ¡la división va primero!
promedio_bien = (nota_corte1 + nota_corte2) / 2

print("Promedio mal calculado: ", promedio_mal)    # 5.5 ¡¿un promedio mayor que las notas?!
print("Promedio bien calculado:", promedio_bien)   # 3.5 ✅

# Este error NO produce ningún mensaje: solo un número equivocado.
# Por eso existe la Regla de Oro.
```

## A4. Estructuras de datos nativas de Python

Un dato solo no sirve de mucho: necesitamos **colecciones**. Python trae cuatro estructuras integradas y cada una tiene su especialidad:

| Estructura | Sintaxis | ¿Ordenada? | ¿Modificable? | ¿Duplicados? | Uso típico en datos |
|---|---|---|---|---|---|
| **Lista** `list` | `[1, 2, 3]` | ✅ | ✅ | ✅ | Secuencia de valores, una columna |
| **Tupla** `tuple` | `(4.6, -74.1)` | ✅ | ❌ inmutable | ✅ | Datos que no deben cambiar (coordenadas) |
| **Diccionario** `dict` | `{"clave": valor}` | ✅ (desde 3.7) | ✅ | ❌ claves únicas | Un registro con campos con nombre |
| **Conjunto** `set` | `{1, 2, 3}` | ❌ | ✅ | ❌ | Valores únicos, comparar grupos |

**En palabras de uso:**
- **Listas:** colecciones modificables, iteración sobre datos, almacenamiento temporal.
- **Tuplas:** datos que no deben cambiar, claves de diccionarios, coordenadas.
- **Diccionarios:** registros estructurados, búsqueda rápida por clave, datos de APIs (JSON).
- **Conjuntos:** eliminar duplicados, pruebas de pertenencia, operaciones de unión/intersección/diferencia.

```python
# LISTAS: la estructura más usada
notas = [3.5, 4.2, 2.8, 5.0, 3.9]

print(notas[0])       # primer elemento (¡se cuenta desde 0!)
print(notas[-1])      # último elemento
print(notas[1:3])     # slicing: desde índice 1 hasta ANTES del 3
print(len(notas))     # cuántos elementos hay
```

```python
# Las listas son modificables (mutables)
notas.append(4.5)        # agregar al final
notas[2] = 3.0           # corregir un valor
notas.sort()             # ordenar EN el lugar
print(notas)
print("Promedio:", sum(notas) / len(notas))   # 👀 paréntesis: Regla de Oro aplicada
print("Máxima:", max(notas), "| Mínima:", min(notas))
```

```python
# TUPLAS: como listas, pero INMUTABLES (no se pueden cambiar)
coordenadas_ean = (4.6482, -74.0631)   # sede El Nogal, Bogotá
print(coordenadas_ean[0])

# Descubre por qué son útiles: intenta modificarla
coordenadas_ean[0] = 5.0   # <- lee el error: las tuplas protegen los datos
```

```python
# DICCIONARIOS: pares clave -> valor. Así se ve UN registro de datos
estudiante = {
    "nombre": "Samuel Rojas",
    "programa": "Ing. de Sistemas",
    "semestre": 4,
    "promedio": 4.1,
    "activo": True
}

print(estudiante["nombre"])          # acceso por clave
print(estudiante.get("correo"))       # .get no falla si la clave no existe
estudiante["correo"] = "srojas@universidadean.edu.co"   # agregar campo
print(estudiante.keys())
print(estudiante.values())
```

```python
# CONJUNTOS (sets): solo valores ÚNICOS. Perfectos para de-duplicar
respuestas = ["sí", "no", "sí", "sí", "no", "tal vez", "sí"]
unicos = set(respuestas)
print(unicos)
print("Opciones distintas:", len(unicos))

# Operaciones de conjuntos: ¿quiénes están en ambos grupos?
curso_python = {"Ana", "Luis", "María", "Juan"}
curso_r = {"María", "Pedro", "Juan"}
print("En ambos cursos:", curso_python & curso_r)
print("Solo en Python:", curso_python - curso_r)
```

```python
# 🧩 LA ESTRUCTURA MÁS IMPORTANTE PARA DATOS:
# una LISTA DE DICCIONARIOS = una tabla (cada dict es una fila)
encuesta = [
    {"nombre": "Ana",   "edad": 20, "programa": "Industrial", "horas_sueno": 7.5},
    {"nombre": "Luis",  "edad": 22, "programa": "Sistemas",   "horas_sueno": 6.0},
    {"nombre": "María", "edad": 21, "programa": "Ambiental",  "horas_sueno": 8.0},
    {"nombre": "Juan",  "edad": 23, "programa": "Sistemas",   "horas_sueno": 5.5},
]

# Recorrerla con un ciclo for:
for persona in encuesta:
    print(persona["nombre"], "duerme", persona["horas_sueno"], "horas")
```

```python
# Ejercicio 3.3 — Rescata la columna
respuestas_edad = ["18", "21", "19 años", "22", "veinte", "20", "23", "N/A"]

# a) Intenta convertir TODAS con int() en un ciclo for... y observa el error
for valor in respuestas_edad:
    print(int(valor))

# b) Convierte SOLO las que se puedan (método .isdigit())
edades_limpias = []
for valor in respuestas_edad:
    if valor.isdigit():
        edades_limpias.append(int(valor))

# c) Imprime cuántas se perdieron y el promedio de las válidas
perdidas = len(respuestas_edad) - len(edades_limpias)
print("Edades perdidas:", perdidas, "de", len(respuestas_edad))
print("Promedio de las válidas:", sum(edades_limpias) / len(edades_limpias))
```

```python
# 🚀 Mi mini-proyecto: registro de mis sesiones de entrenamiento de la semana
entrenamientos = [
    {"dia": "lunes",     "minutos": 45,  "km": 5.2,  "al_gimnasio": True,  "fecha": "2026-07-27"},
    {"dia": "martes",    "minutos": 30,  "km": 3.0,  "al_gimnasio": False, "fecha": "2026-07-28"},
    {"dia": "miércoles", "minutos": 60,  "km": 7.5,  "al_gimnasio": True,  "fecha": "2026-07-29"},
    {"dia": "jueves",    "minutos": 0,   "km": 0.0,  "al_gimnasio": False, "fecha": "2026-07-30"},
    {"dia": "viernes",   "minutos": 50,  "km": 6.1,  "al_gimnasio": True,  "fecha": "2026-07-31"},
    {"dia": "sábado",    "minutos": 90,  "km": 12.0, "al_gimnasio": True,  "fecha": "2026-08-01"},
    {"dia": "domingo",   "minutos": 20,  "km": 2.5,  "al_gimnasio": False, "fecha": "2026-08-02"},
    {"dia": "lunes",     "minutos": 40,  "km": 4.8,  "al_gimnasio": True,  "fecha": "2026-08-03"},
]
```

```python
# 🚀 Tus 3 resultados (promedio, porcentaje y comparación)

# 1) PROMEDIO: minutos promedio de entrenamiento por día
total_minutos = 0
for sesion in entrenamientos:
    total_minutos = total_minutos + sesion["minutos"]
promedio_minutos = total_minutos / len(entrenamientos)
print("Promedio de minutos por día:", promedio_minutos)

# 2) PORCENTAJE: días que fui al gimnasio sobre el total
dias_gimnasio = 0
for sesion in entrenamientos:
    if sesion["al_gimnasio"]:
        dias_gimnasio = dias_gimnasio + 1
porcentaje_gimnasio = (dias_gimnasio / len(entrenamientos)) * 100
print("Días de gimnasio:", dias_gimnasio, "de", len(entrenamientos), f"({porcentaje_gimnasio:.0f}%)")

# 3) COMPARACIÓN: ¿entrené al menos la mitad de los días?
print("¿Entrené al menos la mitad de los días?", dias_gimnasio >= len(entrenamientos) / 2)
```

## ⭐ RETO ESTRELLA (bonus — para valientes)

**La beca DataXperience.** La universidad otorga una beca si el estudiante cumple **todas** estas condiciones:
- Promedio mayor o igual a 4.0, **y**
- estar en semestre 3 o superior, **o** pertenecer a estrato 1 o 2 (cualquiera de las dos), **y**
- **no** tener materias perdidas este semestre.

```python
# ⭐ a) La regla de la beca traducida a UNA expresión booleana
promedio = 4.2
semestre = 2
estrato = 2
materias_perdidas = 0

obtiene_beca = (promedio >= 4.0) and (semestre >= 3 or estrato <= 2) and (materias_perdidas == 0)
print("Caso base (4.2, sem 2, estrato 2, 0 perdidas):", obtiene_beca)

# ⭐ b) Tres casos de prueba
obtiene_beca_caso1 = (4.2 >= 4.0) and (2 >= 3 or 2 <= 2) and (0 == 0)   # True
obtiene_beca_caso2 = (4.5 >= 4.0) and (5 >= 3 or 4 <= 2) and (1 == 0)   # False
obtiene_beca_caso3 = (3.9 >= 4.0) and (6 >= 3 or 1 <= 2) and (0 == 0)   # False
print("Caso 1:", obtiene_beca_caso1)
print("Caso 2:", obtiene_beca_caso2)
print("Caso 3:", obtiene_beca_caso3)

# ⭐ c) Sin paréntesis: evalúa los mismos casos
obtiene_beca_sin_parentesis = promedio >= 4.0 and semestre >= 3 or estrato <= 2 and materias_perdidas == 0
print("Caso base sin paréntesis:", obtiene_beca_sin_parentesis)
# Explicación: sin paréntesis, `and` se agrupa antes que `or`, así que la expresión
# significa (promedio>=4.0 and semestre>=3) OR (estrato<=2 and materias_perdidas==0),
# que NO es la regla pedida: alguien con promedio bajo pero estrato bajo y sin
# materias perdidas podría salir True sin merecer la beca.
```
